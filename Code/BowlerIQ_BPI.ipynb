```python
"""
================================================================================
CRICKET BOWLER PERFORMANCE METRIC - DISCOVERY ANALYSIS
================================================================================

Purpose: Data-driven discovery of what constitutes bowling performance
         across different formats (T20, ODI, Test)

Methodology:
1. Win Contribution Analysis - What helps teams win?
2. Variance Decomposition - What separates great from average?
3. Elite Benchmarking - What makes elite bowlers elite?
4. PCA Analysis - What's the natural combination?
5. Match Situation Analysis - What matters in pressure?

Author: [Your Name]
Date: March 2026
================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Set visualization style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("\n" + "="*80)
print("🔬 CRICKET BOWLER PERFORMANCE METRIC - DISCOVERY ANALYSIS")
print("="*80)

# ============================================================================
# STEP 0: LOAD DATA (MANUAL PATH INPUT)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 0: LOADING DATA")
print("-"*80)

# ============================================================================
# 👇 PROVIDE YOUR FILE PATH HERE
# ============================================================================
FILE_PATH = r"C:\Users\HPP\abt_final_with_all_features.parquet"  # ← CHANGE THIS PATH
# ============================================================================

print(f"\n📂 Loading data from: {FILE_PATH}")

try:
    abt = pd.read_parquet(FILE_PATH)
    print(f"✅ Data loaded successfully!")
    print(f"   Rows: {len(abt):,}")
    print(f"   Columns: {len(abt.columns)}")
    print(f"   Date range: {abt['date'].min()} to {abt['date'].max()}")
except Exception as e:
    print(f"❌ ERROR loading data: {e}")
    print(f"   Please check the file path and try again.")
    exit()

# ============================================================================
# DATA PREPARATION
# ============================================================================

print("\n" + "-"*80)
print("🔧 DATA PREPARATION")
print("-"*80)

# Key metrics we're analyzing
metrics = ['economy_rate', 'wickets', 'strike_rate', 'dot_ball_percentage']

# Filter to only rows where bowler took at least 1 wicket (strike rate defined)
abt_analysis = abt[abt['wickets'] > 0].copy()

print(f"\n📊 Analysis dataset:")
print(f"   Total instances: {len(abt_analysis):,} (bowlers who took ≥1 wicket)")
print(f"   Formats: {abt_analysis['match_type'].value_counts().to_dict()}")

# Check data availability
print(f"\n📊 Metrics availability:")
for metric in metrics:
    available = abt_analysis[metric].notna().sum()
    pct = (available / len(abt_analysis)) * 100
    print(f"   {metric:.<30} {available:,} ({pct:.1f}%)")

print("\n✅ Data preparation complete!")

# ============================================================================
# ANALYSIS 1: WIN CONTRIBUTION ANALYSIS ⭐
# ============================================================================

print("\n\n" + "="*80)
print("📊 ANALYSIS 1: WIN CONTRIBUTION ANALYSIS")
print("="*80)
print("\nQuestion: Which bowling metrics help teams WIN?")
print("Method: Compare winning vs losing teams' bowling performance")

# Create win/loss indicator
abt_analysis['team_won'] = (abt_analysis['bowling_team'] == abt_analysis['winner']).astype(int)

# Remove no-results
abt_win_analysis = abt_analysis[abt_analysis['result_type'].notna()].copy()

print(f"\n📊 Matches with clear results: {len(abt_win_analysis):,}")
print(f"   Bowling team won: {abt_win_analysis['team_won'].sum():,}")
print(f"   Bowling team lost: {(1-abt_win_analysis['team_won']).sum():,}")

# Analysis by format
win_correlations = {}

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - Win Contribution Analysis")
    print(f"{'─'*80}")
    
    fmt_data = abt_win_analysis[abt_win_analysis['match_type'] == fmt].copy()
    
    if len(fmt_data) < 100:
        print(f"   ⚠️ Insufficient data ({len(fmt_data)} instances), skipping...")
        continue
    
    print(f"\n   Analyzing {len(fmt_data):,} bowling performances...")
    
    correlations = {}
    
    for metric in metrics:
        # Calculate correlation with winning
        # For economy/strike_rate: lower is better, so negate
        if metric in ['economy_rate', 'strike_rate']:
            corr = fmt_data[metric].corr(-fmt_data['team_won'] + 1)  # Invert so negative corr = good
        else:  # wickets, dot_ball_pct: higher is better
            corr = fmt_data[metric].corr(fmt_data['team_won'])
        
        correlations[metric] = corr
    
    # Store results
    win_correlations[fmt] = correlations
    
    # Print results
    print(f"\n   📊 Correlation with WINNING (higher = more important):")
    sorted_corr = sorted(correlations.items(), key=lambda x: abs(x[1]), reverse=True)
    for i, (metric, corr) in enumerate(sorted_corr, 1):
        strength = "🔥 STRONG" if abs(corr) > 0.3 else "💪 MEDIUM" if abs(corr) > 0.2 else "📍 WEAK"
        print(f"      {i}. {metric:.<30} {corr:+.4f}  {strength}")
    
    # Statistical significance
    print(f"\n   📊 Mean values: Winning vs Losing teams")
    for metric in metrics:
        win_mean = fmt_data[fmt_data['team_won'] == 1][metric].mean()
        loss_mean = fmt_data[fmt_data['team_won'] == 0][metric].mean()
        diff = win_mean - loss_mean
        
        # T-test
        win_vals = fmt_data[fmt_data['team_won'] == 1][metric].dropna()
        loss_vals = fmt_data[fmt_data['team_won'] == 0][metric].dropna()
        t_stat, p_val = stats.ttest_ind(win_vals, loss_vals)
        
        sig = "***" if p_val < 0.001 else "**" if p_val < 0.01 else "*" if p_val < 0.05 else "ns"
        
        print(f"      {metric:.<30} Win: {win_mean:6.2f}  |  Loss: {loss_mean:6.2f}  |  Diff: {diff:+6.2f}  {sig}")

# Summary visualization
print(f"\n{'─'*80}")
print(f"📊 SUMMARY: Win Contribution Strength by Format")
print(f"{'─'*80}")

fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Analysis 1: Win Contribution by Format\n(Which metrics predict wins?)', 
             fontsize=14, fontweight='bold')

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    if fmt not in win_correlations:
        continue
    
    ax = axes[idx]
    corrs = win_correlations[fmt]
    
    metrics_list = list(corrs.keys())
    values = [abs(corrs[m]) for m in metrics_list]
    colors = ['#e74c3c' if v > 0.3 else '#f39c12' if v > 0.2 else '#95a5a6' for v in values]
    
    bars = ax.barh(metrics_list, values, color=colors, edgecolor='black', alpha=0.8)
    ax.set_xlabel('|Correlation with Win|', fontsize=11)
    ax.set_title(f'{fmt}', fontsize=12, fontweight='bold')
    ax.set_xlim(0, max(values) * 1.2)
    ax.grid(True, alpha=0.3, axis='x')
    
    # Add value labels
    for i, (metric, val) in enumerate(zip(metrics_list, values)):
        ax.text(val + 0.01, i, f'{val:.3f}', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('analysis1_win_contribution.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: analysis1_win_contribution.png")
plt.show()

# ============================================================================
# ANALYSIS 2: VARIANCE DECOMPOSITION 📈
# ============================================================================

print("\n\n" + "="*80)
print("📊 ANALYSIS 2: VARIANCE DECOMPOSITION")
print("="*80)
print("\nQuestion: What separates GREAT from AVERAGE performances?")
print("Method: Coefficient of Variation (CV) - higher CV = more discriminating")

variance_analysis = {}

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - Variance Analysis")
    print(f"{'─'*80}")
    
    fmt_data = abt_analysis[abt_analysis['match_type'] == fmt].copy()
    
    if len(fmt_data) < 100:
        print(f"   ⚠️ Insufficient data, skipping...")
        continue
    
    print(f"\n   Analyzing {len(fmt_data):,} performances...")
    
    cv_dict = {}
    
    print(f"\n   📊 Coefficient of Variation (CV = std/mean):")
    print(f"      Higher CV = metric varies more = better discriminator")
    print()
    
    for metric in metrics:
        data = fmt_data[metric].dropna()
        mean_val = data.mean()
        std_val = data.std()
        cv = std_val / mean_val if mean_val != 0 else 0
        
        cv_dict[metric] = {
            'mean': mean_val,
            'std': std_val,
            'cv': cv
        }
        
        strength = "🔥 HIGH" if cv > 0.4 else "💪 MEDIUM" if cv > 0.25 else "📍 LOW"
        
        print(f"      {metric:.<30}")
        print(f"         Mean: {mean_val:7.2f}  |  Std: {std_val:6.2f}  |  CV: {cv:.3f}  {strength}")
    
    variance_analysis[fmt] = cv_dict
    
    # Interpret
    print(f"\n   💡 INTERPRETATION:")
    sorted_cv = sorted(cv_dict.items(), key=lambda x: x[1]['cv'], reverse=True)
    print(f"      Top discriminator: {sorted_cv[0][0]} (CV={sorted_cv[0][1]['cv']:.3f})")
    print(f"      → This metric shows the MOST variation in {fmt}")
    print(f"      → Best for separating good vs bad performances")

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Analysis 2: Variance Decomposition by Format\n(Which metrics vary most?)', 
             fontsize=14, fontweight='bold')

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    if fmt not in variance_analysis:
        continue
    
    ax = axes[idx]
    cv_data = variance_analysis[fmt]
    
    metrics_list = list(cv_data.keys())
    cv_values = [cv_data[m]['cv'] for m in metrics_list]
    colors = ['#e74c3c' if v > 0.4 else '#f39c12' if v > 0.25 else '#95a5a6' for v in cv_values]
    
    bars = ax.barh(metrics_list, cv_values, color=colors, edgecolor='black', alpha=0.8)
    ax.set_xlabel('Coefficient of Variation', fontsize=11)
    ax.set_title(f'{fmt}', fontsize=12, fontweight='bold')
    ax.set_xlim(0, max(cv_values) * 1.2)
    ax.grid(True, alpha=0.3, axis='x')
    
    for i, (metric, val) in enumerate(zip(metrics_list, cv_values)):
        ax.text(val + 0.01, i, f'{val:.3f}', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('analysis2_variance_decomposition.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: analysis2_variance_decomposition.png")
plt.show()

# ============================================================================
# ANALYSIS 3: ELITE BENCHMARKING 🏆
# ============================================================================

print("\n\n" + "="*80)
print("📊 ANALYSIS 3: ELITE BENCHMARKING")
print("="*80)
print("\nQuestion: What makes ELITE bowlers elite?")
print("Method: Compare top 10% vs median performers")

elite_analysis = {}

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - Elite Benchmarking")
    print(f"{'─'*80}")
    
    fmt_data = abt_analysis[abt_analysis['match_type'] == fmt].copy()
    
    if len(fmt_data) < 100:
        print(f"   ⚠️ Insufficient data, skipping...")
        continue
    
    # Define elite based on career performance
    # Elite = top 10% in both wickets AND economy (format-specific)
    career_data = fmt_data.groupby('bowler').agg({
        'wickets': 'sum',
        'economy_rate': 'mean',
        'matches_at_venue': 'count'  # Using as proxy for matches
    }).reset_index()
    
    # Filter: bowlers with at least 10 matches in format
    career_data = career_data[career_data['matches_at_venue'] >= 10]
    
    # Elite: top 10% wickets AND bottom 25% economy
    wicket_threshold = career_data['wickets'].quantile(0.90)
    economy_threshold = career_data['economy_rate'].quantile(0.25)  # Lower is better
    
    elite_bowlers = career_data[
        (career_data['wickets'] >= wicket_threshold) & 
        (career_data['economy_rate'] <= economy_threshold)
    ]['bowler'].tolist()
    
    # Average: 40-60th percentile
    wicket_avg_low = career_data['wickets'].quantile(0.40)
    wicket_avg_high = career_data['wickets'].quantile(0.60)
    economy_avg_low = career_data['economy_rate'].quantile(0.40)
    economy_avg_high = career_data['economy_rate'].quantile(0.60)
    
    avg_bowlers = career_data[
        (career_data['wickets'] >= wicket_avg_low) & 
        (career_data['wickets'] <= wicket_avg_high) &
        (career_data['economy_rate'] >= economy_avg_low) &
        (career_data['economy_rate'] <= economy_avg_high)
    ]['bowler'].tolist()
    
    print(f"\n   📊 Bowler categories:")
    print(f"      Elite bowlers: {len(elite_bowlers)} (top 10% wickets + economy)")
    print(f"      Average bowlers: {len(avg_bowlers)} (40-60th percentile)")
    
    if len(elite_bowlers) < 5 or len(avg_bowlers) < 10:
        print(f"   ⚠️ Insufficient elite/average bowlers, skipping...")
        continue
    
    # Get match-level performance for these bowlers
    elite_perf = fmt_data[fmt_data['bowler'].isin(elite_bowlers)]
    avg_perf = fmt_data[fmt_data['bowler'].isin(avg_bowlers)]
    
    print(f"\n   📊 Match-level performances:")
    print(f"      Elite: {len(elite_perf):,} performances")
    print(f"      Average: {len(avg_perf):,} performances")
    
    # Compare metrics
    print(f"\n   📊 ELITE vs AVERAGE comparison:")
    print(f"      {'Metric':<25} {'Elite':>10} {'Average':>10} {'Gap':>10} {'Gap%':>10}")
    print(f"      {'-'*70}")
    
    gaps = {}
    
    for metric in metrics:
        elite_mean = elite_perf[metric].mean()
        avg_mean = avg_perf[metric].mean()
        gap = elite_mean - avg_mean
        gap_pct = (gap / avg_mean * 100) if avg_mean != 0 else 0
        
        gaps[metric] = {
            'elite': elite_mean,
            'average': avg_mean,
            'gap': gap,
            'gap_pct': gap_pct
        }
        
        print(f"      {metric:<25} {elite_mean:>10.2f} {avg_mean:>10.2f} {gap:>+10.2f} {gap_pct:>+9.1f}%")
    
    elite_analysis[fmt] = gaps
    
    # Interpret
    print(f"\n   💡 INTERPRETATION:")
    sorted_gaps = sorted(gaps.items(), key=lambda x: abs(x[1]['gap_pct']), reverse=True)
    print(f"      Biggest differentiator: {sorted_gaps[0][0]}")
    print(f"         Gap: {sorted_gaps[0][1]['gap']:+.2f} ({sorted_gaps[0][1]['gap_pct']:+.1f}%)")
    print(f"      → Elite {fmt} bowlers excel MOST in this metric")

# Visualization
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Analysis 3: Elite Benchmarking\n(What makes elite bowlers elite?)', 
             fontsize=14, fontweight='bold')

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    if fmt not in elite_analysis:
        continue
    
    # Absolute gaps
    ax = axes[0, idx]
    gaps_data = elite_analysis[fmt]
    
    metrics_list = list(gaps_data.keys())
    gap_values = [gaps_data[m]['gap'] for m in metrics_list]
    colors = ['#27ae60' if v < 0 else '#e74c3c' for v in gap_values]  # Green if elite is lower (good for economy/strike)
    
    bars = ax.barh(metrics_list, gap_values, color=colors, edgecolor='black', alpha=0.8)
    ax.set_xlabel('Elite - Average', fontsize=11)
    ax.set_title(f'{fmt} (Absolute Gap)', fontsize=12, fontweight='bold')
    ax.axvline(x=0, color='black', linestyle='-', linewidth=0.8)
    ax.grid(True, alpha=0.3, axis='x')
    
    # Percentage gaps
    ax = axes[1, idx]
    gap_pct_values = [abs(gaps_data[m]['gap_pct']) for m in metrics_list]
    colors = ['#e74c3c' if v > 15 else '#f39c12' if v > 10 else '#95a5a6' for v in gap_pct_values]
    
    bars = ax.barh(metrics_list, gap_pct_values, color=colors, edgecolor='black', alpha=0.8)
    ax.set_xlabel('|% Difference|', fontsize=11)
    ax.set_title(f'{fmt} (% Gap)', fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='x')
    
    for i, (metric, val) in enumerate(zip(metrics_list, gap_pct_values)):
        ax.text(val + 0.5, i, f'{val:.1f}%', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('analysis3_elite_benchmarking.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: analysis3_elite_benchmarking.png")
plt.show()

# ============================================================================
# ANALYSIS 4: PCA - NATURAL COMBINATION 🧮
# ============================================================================

print("\n\n" + "="*80)
print("📊 ANALYSIS 4: PRINCIPAL COMPONENT ANALYSIS")
print("="*80)
print("\nQuestion: What's the NATURAL combination that explains performance?")
print("Method: PCA to find mathematically optimal combination")

pca_analysis = {}

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - PCA Analysis")
    print(f"{'─'*80}")
    
    fmt_data = abt_analysis[abt_analysis['match_type'] == fmt].copy()
    
    if len(fmt_data) < 100:
        print(f"   ⚠️ Insufficient data, skipping...")
        continue
    
    # Prepare data for PCA
    pca_data = fmt_data[metrics].dropna()
    
    print(f"\n   📊 Analyzing {len(pca_data):,} complete performances...")
    
    # Standardize (important for PCA)
    scaler = StandardScaler()
    pca_scaled = scaler.fit_transform(pca_data)
    
    # Run PCA
    pca = PCA()
    pca_transformed = pca.fit_transform(pca_scaled)
    
    # Get results
    explained_var = pca.explained_variance_ratio_
    loadings = pca.components_
    
    print(f"\n   📊 Variance explained by components:")
    for i, var in enumerate(explained_var, 1):
        print(f"      PC{i}: {var*100:5.1f}%  (cumulative: {explained_var[:i].sum()*100:.1f}%)")
    
    print(f"\n   📊 PC1 Loadings (Natural weights):")
    pc1_loadings = loadings[0]
    
    loading_dict = {}
    for metric, loading in zip(metrics, pc1_loadings):
        loading_dict[metric] = loading
        strength = "🔥 STRONG" if abs(loading) > 0.5 else "💪 MEDIUM" if abs(loading) > 0.3 else "📍 WEAK"
        sign = "↑ Positive" if loading > 0 else "↓ Negative"
        print(f"      {metric:.<30} {loading:+.4f}  {strength}  ({sign})")
    
    pca_analysis[fmt] = {
        'explained_var': explained_var,
        'pc1_loadings': loading_dict,
        'pc1_var': explained_var[0]
    }
    
    print(f"\n   💡 INTERPRETATION:")
    sorted_loadings = sorted(loading_dict.items(), key=lambda x: abs(x[1]), reverse=True)
    print(f"      PC1 explains {explained_var[0]*100:.1f}% of variance")
    print(f"      Strongest component: {sorted_loadings[0][0]} (loading={sorted_loadings[0][1]:+.3f})")
    print(f"      → This metric naturally dominates {fmt} performance variation")

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Analysis 4: PCA - Natural Combination of Metrics\n(Mathematically optimal weights)', 
             fontsize=14, fontweight='bold')

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    if fmt not in pca_analysis:
        continue
    
    ax = axes[idx]
    loadings = pca_analysis[fmt]['pc1_loadings']
    var_explained = pca_analysis[fmt]['pc1_var']
    
    metrics_list = list(loadings.keys())
    loading_values = [abs(loadings[m]) for m in metrics_list]
    colors = ['#e74c3c' if v > 0.5 else '#f39c12' if v > 0.3 else '#95a5a6' for v in loading_values]
    
    bars = ax.barh(metrics_list, loading_values, color=colors, edgecolor='black', alpha=0.8)
    ax.set_xlabel('|PC1 Loading|', fontsize=11)
    ax.set_title(f'{fmt}\n(Explains {var_explained*100:.1f}% variance)', 
                 fontsize=12, fontweight='bold')
    ax.set_xlim(0, max(loading_values) * 1.2)
    ax.grid(True, alpha=0.3, axis='x')
    
    for i, (metric, val) in enumerate(zip(metrics_list, loading_values)):
        ax.text(val + 0.01, i, f'{val:.3f}', va='center', fontsize=9)

plt.tight_layout()
plt.savefig('analysis4_pca_analysis.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: analysis4_pca_analysis.png")
plt.show()

# ============================================================================
# ANALYSIS 5: MATCH SITUATION ANALYSIS 🎯
# ============================================================================

print("\n\n" + "="*80)
print("📊 ANALYSIS 5: MATCH SITUATION ANALYSIS")
print("="*80)
print("\nQuestion: What matters in HIGH-PRESSURE situations?")
print("Method: Compare close matches vs blowouts")

situation_analysis = {}

# Define close matches
for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - Situation Analysis")
    print(f"{'─'*80}")
    
    fmt_data = abt_win_analysis[abt_win_analysis['match_type'] == fmt].copy()
    
    if len(fmt_data) < 100:
        print(f"   ⚠️ Insufficient data, skipping...")
        continue
    
    # Define close matches based on win margin
    if 'win_by_runs' in fmt_data.columns and 'win_by_wickets' in fmt_data.columns:
        # Close match criteria
        runs_margin_threshold = 20 if fmt == 'T20' else 30 if fmt == 'ODI' else 50
        wickets_margin_threshold = 3
        
        fmt_data['is_close'] = (
            ((fmt_data['win_by_runs'].fillna(999) <= runs_margin_threshold) & (fmt_data['win_by_runs'].notna())) |
            ((fmt_data['win_by_wickets'].fillna(999) <= wickets_margin_threshold) & (fmt_data['win_by_wickets'].notna()))
        ).astype(int)
        
        close_matches = fmt_data[fmt_data['is_close'] == 1]
        blowout_matches = fmt_data[fmt_data['is_close'] == 0]
        
        print(f"\n   📊 Match categories:")
        print(f"      Close matches: {len(close_matches):,} ({len(close_matches)/len(fmt_data)*100:.1f}%)")
        print(f"      Blowouts: {len(blowout_matches):,} ({len(blowout_matches)/len(fmt_data)*100:.1f}%)")
        
        if len(close_matches) < 50:
            print(f"   ⚠️ Insufficient close matches, skipping...")
            continue
        
        # Win correlation in close matches vs blowouts
        print(f"\n   📊 Win correlation in CLOSE matches (pressure!):")
        close_corr = {}
        for metric in metrics:
            if metric in ['economy_rate', 'strike_rate']:
                corr = close_matches[metric].corr(-close_matches['team_won'] + 1)
            else:
                corr = close_matches[metric].corr(close_matches['team_won'])
            close_corr[metric] = corr
            strength = "🔥 CRITICAL" if abs(corr) > 0.35 else "💪 IMPORTANT" if abs(corr) > 0.25 else "📍 MODERATE"
            print(f"      {metric:.<30} {corr:+.4f}  {strength}")
        
        print(f"\n   📊 Win correlation in BLOWOUTS:")
        blowout_corr = {}
        for metric in metrics:
            if metric in ['economy_rate', 'strike_rate']:
                corr = blowout_matches[metric].corr(-blowout_matches['team_won'] + 1)
            else:
                corr = blowout_matches[metric].corr(blowout_matches['team_won'])
            blowout_corr[metric] = corr
            print(f"      {metric:.<30} {corr:+.4f}")
        
        situation_analysis[fmt] = {
            'close': close_corr,
            'blowout': blowout_corr
        }
        
        print(f"\n   💡 INTERPRETATION:")
        sorted_close = sorted(close_corr.items(), key=lambda x: abs(x[1]), reverse=True)
        print(f"      In PRESSURE situations: {sorted_close[0][0]} matters most")
        print(f"         (correlation: {sorted_close[0][1]:+.3f})")
        print(f"      → This is what separates winners from losers in tight games")

# Visualization
if len(situation_analysis) > 0:
    fig, axes = plt.subplots(1, min(3, len(situation_analysis)), 
                             figsize=(6*min(3, len(situation_analysis)), 6))
    fig.suptitle('Analysis 5: Match Situation Analysis\n(Close matches vs Blowouts)', 
                 fontsize=14, fontweight='bold')
    
    if len(situation_analysis) == 1:
        axes = [axes]
    
    for idx, (fmt, data) in enumerate(situation_analysis.items()):
        ax = axes[idx]
        
        close_corr = data['close']
        blowout_corr = data['blowout']
        
        x = np.arange(len(metrics))
        width = 0.35
        
        close_vals = [abs(close_corr[m]) for m in metrics]
        blowout_vals = [abs(blowout_corr[m]) for m in metrics]
        
        ax.bar(x - width/2, close_vals, width, label='Close Matches', 
               color='#e74c3c', edgecolor='black', alpha=0.8)
        ax.bar(x + width/2, blowout_vals, width, label='Blowouts',
               color='#95a5a6', edgecolor='black', alpha=0.8)
        
        ax.set_ylabel('|Correlation with Win|', fontsize=11)
        ax.set_title(f'{fmt}', fontsize=12, fontweight='bold')
        ax.set_xticks(x)
        ax.set_xticklabels([m.replace('_', '\n') for m in metrics], fontsize=9)
        ax.legend()
        ax.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.savefig('analysis5_situation_analysis.png', dpi=300, bbox_inches='tight')
    print(f"\n💾 Saved: analysis5_situation_analysis.png")
    plt.show()

# ============================================================================
# FINAL SYNTHESIS: RECOMMENDED WEIGHTS
# ============================================================================

print("\n\n" + "="*80)
print("🎯 FINAL SYNTHESIS: RECOMMENDED PERFORMANCE METRIC WEIGHTS")
print("="*80)

print("\n📊 Synthesizing findings from all 5 analyses...")

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'='*80}")
    print(f"🏏 {fmt.upper()} FORMAT - RECOMMENDED WEIGHTS")
    print(f"{'='*80}")
    
    # Collect evidence from all analyses
    evidence = {}
    
    for metric in metrics:
        scores = []
        reasons = []
        
        # Evidence 1: Win contribution
        if fmt in win_correlations:
            corr = abs(win_correlations[fmt].get(metric, 0))
            scores.append(corr)
            reasons.append(f"Win corr: {corr:.3f}")
        
        # Evidence 2: Variance
        if fmt in variance_analysis:
            cv = variance_analysis[fmt].get(metric, {}).get('cv', 0)
            # Normalize CV to 0-1 scale
            cv_norm = min(cv / 0.5, 1.0)  # Cap at 0.5 = 1.0
            scores.append(cv_norm)
            reasons.append(f"CV: {cv:.3f}")
        
        # Evidence 3: Elite gap
        if fmt in elite_analysis:
            gap_pct = abs(elite_analysis[fmt].get(metric, {}).get('gap_pct', 0))
            # Normalize to 0-1 scale
            gap_norm = min(gap_pct / 30, 1.0)  # 30% gap = 1.0
            scores.append(gap_norm)
            reasons.append(f"Elite gap: {gap_pct:.1f}%")
        
        # Evidence 4: PCA loading
        if fmt in pca_analysis:
            loading = abs(pca_analysis[fmt]['pc1_loadings'].get(metric, 0))
            scores.append(loading)
            reasons.append(f"PCA: {loading:.3f}")
        
        # Evidence 5: Pressure correlation
        if fmt in situation_analysis:
            pressure_corr = abs(situation_analysis[fmt]['close'].get(metric, 0))
            scores.append(pressure_corr)
            reasons.append(f"Pressure: {pressure_corr:.3f}")
        
        evidence[metric] = {
            'scores': scores,
            'mean_score': np.mean(scores) if scores else 0,
            'reasons': reasons
        }
    
    # Calculate normalized weights
    total_score = sum([e['mean_score'] for e in evidence.values()])
    
    print(f"\n📊 EVIDENCE-BASED WEIGHTS:")
    print(f"   {'Metric':<25} {'Weight':>8} {'Evidence'}")
    print(f"   {'-'*70}")
    
    for metric in sorted(evidence.keys(), key=lambda x: evidence[x]['mean_score'], reverse=True):
        weight = evidence[metric]['mean_score'] / total_score if total_score > 0 else 0
        evidence_str = " | ".join(evidence[metric]['reasons'])
        print(f"   {metric:<25} {weight:>7.1%}   {evidence_str}")
    
    print(f"\n   💡 JUSTIFICATION:")
    sorted_metrics = sorted(evidence.items(), key=lambda x: x[1]['mean_score'], reverse=True)
    print(f"      Highest weight: {sorted_metrics[0][0]} ({sorted_metrics[0][1]['mean_score']/total_score:.1%})")
    print(f"      → Supported by {len(sorted_metrics[0][1]['scores'])} converging analyses")
    print(f"      → {', '.join(sorted_metrics[0][1]['reasons'])}")

# ============================================================================
# SAVE RESULTS
# ============================================================================

print("\n\n" + "="*80)
print("💾 SAVING RESULTS")
print("="*80)

results_summary = {
    'win_correlations': win_correlations,
    'variance_analysis': variance_analysis,
    'elite_analysis': elite_analysis,
    'pca_analysis': pca_analysis,
    'situation_analysis': situation_analysis
}

import pickle
with open('performance_metric_discovery_results.pkl', 'wb') as f:
    pickle.dump(results_summary, f)

print(f"\n✅ All results saved to: performance_metric_discovery_results.pkl")

print(f"\n📁 Visualizations created:")
print(f"   - analysis1_win_contribution.png")
print(f"   - analysis2_variance_decomposition.png")
print(f"   - analysis3_elite_benchmarking.png")
print(f"   - analysis4_pca_analysis.png")
print(f"   - analysis5_situation_analysis.png")

print("\n\n" + "="*80)
print("✅ DISCOVERY ANALYSIS COMPLETE!")
print("="*80)

print("\n🎯 NEXT STEPS:")
print("   1. Review the recommended weights from all 5 analyses")
print("   2. Decide on final weights based on converging evidence")
print("   3. Create your unique Performance Index")
print("   4. Validate against known great bowlers")
print("   5. Name it and use it for modeling!")

print("\n" + "="*80)
```

```python
"""
================================================================================
THE [YOUR_NAME] BOWLING PERFORMANCE INDEX (BPI)
================================================================================

Purpose: Create empirically-validated, format-specific bowling performance metric
Based on: 5 independent statistical analyses on 49,945 bowling performances

Methodology:
- Normalize each component metric to 0-100 scale
- Apply format-specific weights from discovery analysis
- Validate against cricket legends

Author: [Your Name]
Date: March 2026
================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("\n" + "="*80)
print("🏏 THE BOWLING PERFORMANCE INDEX (BPI)")
print("="*80)

# ============================================================================
# STEP 0: LOAD DATA
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 0: LOADING DATA")
print("-"*80)

# 👇 CHANGE THIS PATH
FILE_PATH = 'C:/Users/HPP/abt_final_with_all_features.parquet'

abt = pd.read_parquet(FILE_PATH)
print(f"✅ Loaded: {len(abt):,} rows")

# Filter to performances with at least 1 wicket (for strike rate)
abt_bpi = abt[abt['wickets'] > 0].copy()
print(f"📊 Analysis dataset: {len(abt_bpi):,} performances (wickets ≥ 1)")

# ============================================================================
# STEP 1: DEFINE FORMAT-SPECIFIC WEIGHTS (FROM DISCOVERY ANALYSIS)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 1: FORMAT-SPECIFIC WEIGHTS (EMPIRICALLY DERIVED)")
print("-"*80)

# Weights from discovery analysis
WEIGHTS = {
    'T20': {
        'economy_rate': 0.287,
        'wickets': 0.248,
        'dot_ball_percentage': 0.245,
        'strike_rate': 0.220
    },
    'ODI': {
        'strike_rate': 0.290,
        'economy_rate': 0.270,
        'wickets': 0.256,
        'dot_ball_percentage': 0.184
    },
    'Test': {
        'wickets': 0.336,
        'strike_rate': 0.279,
        'economy_rate': 0.220,
        'dot_ball_percentage': 0.165
    }
}

print("\n📊 Weights Summary:")
for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n   {fmt.upper()}:")
    for metric, weight in sorted(WEIGHTS[fmt].items(), key=lambda x: x[1], reverse=True):
        print(f"      {metric:.<30} {weight:.1%}")

# ============================================================================
# STEP 2: CREATE NORMALIZED COMPONENT SCORES (0-100 SCALE)
# ============================================================================

print("\n\n" + "-"*80)
print("📊 STEP 2: NORMALIZING COMPONENT SCORES")
print("-"*80)

def normalize_to_100(series, lower_is_better=False):
    """
    Normalize a series to 0-100 scale
    
    Parameters:
    - series: pandas Series to normalize
    - lower_is_better: True for economy/strike_rate, False for wickets/dot%
    
    Returns:
    - Normalized series where 100 = best, 0 = worst
    """
    # Remove outliers (use 1st and 99th percentile as bounds)
    lower_bound = series.quantile(0.01)
    upper_bound = series.quantile(0.99)
    series_clipped = series.clip(lower_bound, upper_bound)
    
    # Normalize to 0-1
    min_val = series_clipped.min()
    max_val = series_clipped.max()
    
    if max_val == min_val:
        return pd.Series(50, index=series.index)  # All same = 50
    
    normalized = (series_clipped - min_val) / (max_val - min_val)
    
    # If lower is better (economy, strike rate), invert
    if lower_is_better:
        normalized = 1 - normalized
    
    # Scale to 0-100
    return normalized * 100

# Create normalized scores for each format
print("\n📊 Creating normalized component scores (0-100 scale)...")
print("   100 = Best performance, 0 = Worst performance")

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n   Processing {fmt}...")
    
    fmt_data = abt_bpi[abt_bpi['match_type'] == fmt].copy()
    
    if len(fmt_data) == 0:
        continue
    
    # Normalize each metric
    # Economy & Strike Rate: Lower is better
    fmt_data[f'economy_score_{fmt}'] = normalize_to_100(
        fmt_data['economy_rate'], 
        lower_is_better=True
    )
    
    fmt_data[f'strike_score_{fmt}'] = normalize_to_100(
        fmt_data['strike_rate'], 
        lower_is_better=True
    )
    
    # Wickets & Dot%: Higher is better
    fmt_data[f'wicket_score_{fmt}'] = normalize_to_100(
        fmt_data['wickets'], 
        lower_is_better=False
    )
    
    fmt_data[f'dot_score_{fmt}'] = normalize_to_100(
        fmt_data['dot_ball_percentage'], 
        lower_is_better=False
    )
    
    # Store back
    abt_bpi.loc[fmt_data.index, f'economy_score_{fmt}'] = fmt_data[f'economy_score_{fmt}']
    abt_bpi.loc[fmt_data.index, f'strike_score_{fmt}'] = fmt_data[f'strike_score_{fmt}']
    abt_bpi.loc[fmt_data.index, f'wicket_score_{fmt}'] = fmt_data[f'wicket_score_{fmt}']
    abt_bpi.loc[fmt_data.index, f'dot_score_{fmt}'] = fmt_data[f'dot_score_{fmt}']
    
    print(f"      ✓ Normalized: economy, strike_rate, wickets, dot_ball_percentage")

print("\n✅ Component normalization complete!")

# ============================================================================
# STEP 3: CALCULATE BOWLING PERFORMANCE INDEX (BPI)
# ============================================================================

print("\n\n" + "-"*80)
print("📊 STEP 3: CALCULATING BOWLING PERFORMANCE INDEX (BPI)")
print("-"*80)

print("\n📊 Formula:")
print("   BPI = w1 × Economy_Score + w2 × Wicket_Score + ")
print("         w3 × Dot_Score + w4 × Strike_Score")
print("\n   Where all scores are normalized 0-100 and weights sum to 1.0")

def calculate_bpi(row):
    """Calculate BPI for a single performance"""
    fmt = row['match_type']
    
    if fmt not in ['T20', 'ODI', 'Test']:
        return np.nan
    
    # Get component scores
    economy_score = row[f'economy_score_{fmt}']
    strike_score = row[f'strike_score_{fmt}']
    wicket_score = row[f'wicket_score_{fmt}']
    dot_score = row[f'dot_score_{fmt}']
    
    # Check for missing values
    if pd.isna([economy_score, strike_score, wicket_score, dot_score]).any():
        return np.nan
    
    # Get weights
    weights = WEIGHTS[fmt]
    
    # Calculate BPI
    bpi = (
        weights['economy_rate'] * economy_score +
        weights['strike_rate'] * strike_score +
        weights['wickets'] * wicket_score +
        weights['dot_ball_percentage'] * dot_score
    )
    
    return bpi

# Calculate BPI for all performances
print("\n📊 Calculating BPI for all performances...")
abt_bpi['BPI'] = abt_bpi.apply(calculate_bpi, axis=1)

# Remove rows with missing BPI
abt_bpi = abt_bpi[abt_bpi['BPI'].notna()].copy()

print(f"✅ BPI calculated for {len(abt_bpi):,} performances")

# ============================================================================
# STEP 4: BPI STATISTICS & INTERPRETATION
# ============================================================================

print("\n\n" + "-"*80)
print("📊 STEP 4: BPI STATISTICS")
print("-"*80)

print(f"\n📊 Overall BPI Distribution:")
print(f"   Mean: {abt_bpi['BPI'].mean():.2f}")
print(f"   Median: {abt_bpi['BPI'].median():.2f}")
print(f"   Std Dev: {abt_bpi['BPI'].std():.2f}")
print(f"   Min: {abt_bpi['BPI'].min():.2f}")
print(f"   Max: {abt_bpi['BPI'].max():.2f}")

print(f"\n📊 BPI by Format:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_bpi[abt_bpi['match_type'] == fmt]['BPI']
    if len(fmt_bpi) > 0:
        print(f"   {fmt}: Mean={fmt_bpi.mean():.2f}, Median={fmt_bpi.median():.2f}, "
              f"Std={fmt_bpi.std():.2f}")

# Define performance categories
percentiles = abt_bpi['BPI'].quantile([0.1, 0.25, 0.5, 0.75, 0.9])

print(f"\n📊 Performance Tiers (BPI ranges):")
print(f"   Elite (Top 10%): BPI > {percentiles[0.9]:.2f}")
print(f"   Excellent (75-90%): BPI {percentiles[0.75]:.2f} - {percentiles[0.9]:.2f}")
print(f"   Good (50-75%): BPI {percentiles[0.5]:.2f} - {percentiles[0.75]:.2f}")
print(f"   Average (25-50%): BPI {percentiles[0.25]:.2f} - {percentiles[0.5]:.2f}")
print(f"   Below Average (10-25%): BPI {percentiles[0.1]:.2f} - {percentiles[0.25]:.2f}")
print(f"   Poor (Bottom 10%): BPI < {percentiles[0.1]:.2f}")

# Visualize BPI distribution
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Bowling Performance Index (BPI) - Distribution Analysis', 
             fontsize=16, fontweight='bold')

# Overall distribution
ax = axes[0, 0]
ax.hist(abt_bpi['BPI'], bins=50, edgecolor='black', alpha=0.7, color='steelblue')
ax.axvline(abt_bpi['BPI'].mean(), color='red', linestyle='--', 
           linewidth=2, label=f'Mean: {abt_bpi["BPI"].mean():.1f}')
ax.axvline(abt_bpi['BPI'].median(), color='green', linestyle='--',
           linewidth=2, label=f'Median: {abt_bpi["BPI"].median():.1f}')
ax.set_xlabel('BPI Score', fontsize=12)
ax.set_ylabel('Frequency', fontsize=12)
ax.set_title('Overall BPI Distribution', fontsize=12, fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

# BPI by format
ax = axes[0, 1]
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_bpi[abt_bpi['match_type'] == fmt]['BPI']
    if len(fmt_bpi) > 0:
        ax.hist(fmt_bpi, bins=30, alpha=0.5, label=fmt, edgecolor='black')
ax.set_xlabel('BPI Score', fontsize=12)
ax.set_ylabel('Frequency', fontsize=12)
ax.set_title('BPI Distribution by Format', fontsize=12, fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

# BPI vs Economy (validation)
ax = axes[1, 0]
sample = abt_bpi.sample(min(5000, len(abt_bpi)))
ax.scatter(sample['economy_rate'], sample['BPI'], alpha=0.3, s=20, color='steelblue')
corr = abt_bpi['economy_rate'].corr(abt_bpi['BPI'])
ax.set_xlabel('Economy Rate', fontsize=12)
ax.set_ylabel('BPI', fontsize=12)
ax.set_title(f'BPI vs Economy Rate (r={corr:.3f})', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)

# BPI vs Wickets (validation)
ax = axes[1, 1]
ax.scatter(sample['wickets'], sample['BPI'], alpha=0.3, s=20, color='coral')
corr = abt_bpi['wickets'].corr(abt_bpi['BPI'])
ax.set_xlabel('Wickets', fontsize=12)
ax.set_ylabel('BPI', fontsize=12)
ax.set_title(f'BPI vs Wickets (r={corr:.3f})', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('bpi_distribution_analysis.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: bpi_distribution_analysis.png")
plt.show()

# ============================================================================
# STEP 5: VALIDATE AGAINST CRICKET LEGENDS 🏆
# ============================================================================

print("\n\n" + "="*80)
print("🏆 STEP 5: VALIDATION AGAINST CRICKET LEGENDS")
print("="*80)

print("\n📊 Calculating career BPI for all bowlers...")

# Calculate career average BPI
career_bpi = abt_bpi.groupby(['bowler', 'match_type']).agg({
    'BPI': 'mean',
    'wickets': 'sum',
    'economy_rate': 'mean',
    'strike_rate': 'mean',
    'dot_ball_percentage': 'mean',
    'match_id': 'count'  # Using as proxy for matches
}).reset_index()

career_bpi.columns = ['bowler', 'format', 'Career_BPI', 'Total_Wickets', 
                       'Avg_Economy', 'Avg_Strike_Rate', 'Avg_Dot_Pct', 'Matches']

# Filter: bowlers with at least 20 matches in format (for reliability)
career_bpi = career_bpi[career_bpi['Matches'] >= 20].copy()

print(f"✅ Career BPI calculated for {len(career_bpi)} bowler-format combinations")
print(f"   (Minimum 20 matches per format)")

# Known legends to validate (adjust names as they appear in your data)
LEGENDS = {
    'T20': ['jj bumrah', 'rashid khan', 'sunil narine', 'lasith malinga', 
            'sl malinga', 'r ashwin'],
    'ODI': ['m muralitharan', 'wasim akram', 'glenn mcgrath', 'shahid afridi',
            'jj bumrah', 'mitchell starc'],
    'Test': ['m muralitharan', 'shane warne', 'anil kumble', 'james anderson',
             'r ashwin', 'dale steyn']
}

print("\n" + "="*80)
print("🏆 LEGEND VALIDATION")
print("="*80)

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - TOP 20 BOWLERS BY BPI")
    print(f"{'─'*80}")
    
    fmt_data = career_bpi[career_bpi['format'] == fmt].copy()
    
    if len(fmt_data) == 0:
        print(f"   No data available")
        continue
    
    # Sort by Career BPI
    top_20 = fmt_data.nlargest(20, 'Career_BPI')
    
    print(f"\n{'Rank':<6} {'Bowler':<25} {'BPI':<8} {'Matches':<10} {'Wickets':<10} {'Economy':<10}")
    print(f"{'-'*80}")
    
    for idx, (i, row) in enumerate(top_20.iterrows(), 1):
        is_legend = any(legend in row['bowler'].lower() for legend in LEGENDS.get(fmt, []))
        marker = "⭐" if is_legend else "  "
        
        print(f"{idx:<6} {row['bowler'][:24]:<25} {row['Career_BPI']:<8.2f} "
              f"{int(row['Matches']):<10} {int(row['Total_Wickets']):<10} "
              f"{row['Avg_Economy']:<10.2f} {marker}")
    
    # Check if legends are in top rankings
    print(f"\n   📊 Legend Check:")
    legends_in_top = []
    for legend in LEGENDS.get(fmt, []):
        legend_data = fmt_data[fmt_data['bowler'].str.contains(legend, case=False, na=False)]
        if len(legend_data) > 0:
            best_match = legend_data.nlargest(1, 'Career_BPI').iloc[0]
            rank = (fmt_data['Career_BPI'] > best_match['Career_BPI']).sum() + 1
            legends_in_top.append((legend, rank, best_match['Career_BPI']))
    
    if legends_in_top:
        for legend, rank, bpi in sorted(legends_in_top, key=lambda x: x[1]):
            status = "✅ TOP 10" if rank <= 10 else "✅ TOP 20" if rank <= 20 else "⚠️  LOWER"
            print(f"      {legend:.<30} Rank: {rank:>3}  BPI: {bpi:>6.2f}  {status}")

# ============================================================================
# VISUALIZATION: ELITE VS AVERAGE COMPARISON
# ============================================================================

print("\n\n" + "-"*80)
print("📊 CREATING ELITE VS AVERAGE COMPARISON")
print("-"*80)

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('BPI Validation: Elite vs Average Bowlers', 
             fontsize=16, fontweight='bold')

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    fmt_data = career_bpi[career_bpi['format'] == fmt].copy()
    
    if len(fmt_data) == 0:
        continue
    
    # Define elite (top 10%) and average (40-60%)
    elite_threshold = fmt_data['Career_BPI'].quantile(0.90)
    avg_low = fmt_data['Career_BPI'].quantile(0.40)
    avg_high = fmt_data['Career_BPI'].quantile(0.60)
    
    elite = fmt_data[fmt_data['Career_BPI'] >= elite_threshold]
    average = fmt_data[(fmt_data['Career_BPI'] >= avg_low) & 
                       (fmt_data['Career_BPI'] <= avg_high)]
    
    # Top bowlers bar chart
    ax = axes[0, idx]
    top_10 = fmt_data.nlargest(10, 'Career_BPI')
    colors = ['#2ecc71' if bpi > elite_threshold else '#3498db' 
              for bpi in top_10['Career_BPI']]
    
    y_pos = range(len(top_10))
    ax.barh(y_pos, top_10['Career_BPI'], color=colors, edgecolor='black', alpha=0.8)
    ax.set_yticks(y_pos)
    ax.set_yticklabels([name[:20] for name in top_10['bowler']], fontsize=9)
    ax.set_xlabel('Career BPI', fontsize=11)
    ax.set_title(f'{fmt} - Top 10 Bowlers', fontsize=12, fontweight='bold')
    ax.invert_yaxis()
    ax.grid(True, alpha=0.3, axis='x')
    
    # Elite vs Average comparison
    ax = axes[1, idx]
    
    metrics = ['Avg_Economy', 'Total_Wickets', 'Avg_Dot_Pct']
    metric_labels = ['Economy', 'Wickets\n(Total)', 'Dot %']
    
    elite_means = [elite[m].mean() for m in metrics]
    avg_means = [average[m].mean() for m in metrics]
    
    # Normalize for visualization
    normalized_elite = []
    normalized_avg = []
    for i, metric in enumerate(metrics):
        max_val = max(elite_means[i], avg_means[i])
        normalized_elite.append(elite_means[i] / max_val * 100)
        normalized_avg.append(avg_means[i] / max_val * 100)
    
    x = np.arange(len(metric_labels))
    width = 0.35
    
    ax.bar(x - width/2, normalized_elite, width, label='Elite (Top 10%)', 
           color='#2ecc71', edgecolor='black', alpha=0.8)
    ax.bar(x + width/2, normalized_avg, width, label='Average (40-60%)',
           color='#95a5a6', edgecolor='black', alpha=0.8)
    
    ax.set_ylabel('Normalized Score', fontsize=11)
    ax.set_title(f'{fmt} - Elite vs Average', fontsize=12, fontweight='bold')
    ax.set_xticks(x)
    ax.set_xticklabels(metric_labels, fontsize=10)
    ax.legend()
    ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('bpi_elite_validation.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: bpi_elite_validation.png")
plt.show()

# ============================================================================
# SAVE ENHANCED ABT WITH BPI
# ============================================================================

print("\n\n" + "-"*80)
print("💾 SAVING ENHANCED ABT")
print("-"*80)

abt_bpi.to_parquet('abt_with_bpi.parquet')
print(f"✅ Saved: abt_with_bpi.parquet ({len(abt_bpi):,} rows)")

career_bpi.to_csv('career_bpi_rankings.csv', index=False)
print(f"✅ Saved: career_bpi_rankings.csv ({len(career_bpi)} bowler-format combinations)")

# ============================================================================
# FINAL SUMMARY
# ============================================================================

print("\n\n" + "="*80)
print("✅ BOWLING PERFORMANCE INDEX (BPI) COMPLETE!")
print("="*80)

print("\n🎯 WHAT YOU NOW HAVE:")
print("   1. Format-specific performance index (0-100 scale)")
print("   2. Empirically-validated weights from 5 analyses")
print("   3. Validation against cricket legends")
print("   4. Career BPI rankings for all bowlers")

print("\n📊 BPI SUMMARY:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_bpi[abt_bpi['match_type'] == fmt]['BPI']
    if len(fmt_bpi) > 0:
        print(f"   {fmt}: Mean={fmt_bpi.mean():.2f}, "
              f"Range=[{fmt_bpi.min():.2f}, {fmt_bpi.max():.2f}]")

print("\n🚀 NEXT STEPS:")
print("   1. Use BPI as target variable in modeling")
print("   2. Predict BPI given match conditions (venue, opponent, etc.)")
print("   3. Feature importance → understand drivers of elite performance")
print("   4. Build format-specific models")

print("\n" + "="*80)
```

```python
"""
================================================================================
BPI REFINEMENT TESTING - T20 FORMAT ONLY
================================================================================

Purpose: Test 5 different fixes + 1 combined approach to improve BPI rankings
Format: T20 only (for faster testing)

Approaches:
1. Higher match threshold (20 → 50 matches)
2. Opposition-adjusted BPI
3. Confidence-weighted BPI
4. Full Member nations only
5. Recency-weighted BPI
6. COMBINED (all fixes together)

================================================================================
"""

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("\n" + "="*80)
print("🧪 BPI REFINEMENT TESTING - T20 FORMAT")
print("="*80)

# ============================================================================
# LOAD DATA
# ============================================================================

print("\n📊 Loading data...")

# 👇 CHANGE PATH
FILE_PATH = 'C:/Users/HPP/abt_with_bpi.parquet'

abt = pd.read_parquet(FILE_PATH)
print(f"✅ Loaded: {len(abt):,} rows")

# Filter to T20 only
t20_data = abt[abt['match_type'] == 'T20'].copy()
print(f"📊 T20 performances: {len(t20_data):,}")

# Calculate career BPI (baseline - 20 match minimum)
career_t20 = t20_data.groupby('bowler').agg({
    'BPI': 'mean',
    'wickets': 'sum',
    'economy_rate': 'mean',
    'strike_rate': 'mean',
    'dot_ball_percentage': 'mean',
    'match_id': 'count',  # Matches
    'date': ['min', 'max'],  # Career span
    'opponent_is_full_member': 'mean',  # % against full members
    'opponent_batting_strength': 'mean'  # Avg opponent quality
}).reset_index()

career_t20.columns = ['bowler', 'Career_BPI', 'Total_Wickets', 'Avg_Economy', 
                       'Avg_Strike_Rate', 'Avg_Dot_Pct', 'Matches',
                       'Career_Start', 'Career_End', 'Pct_vs_FullMembers', 
                       'Avg_Opponent_Strength']

# Baseline: 20 match minimum
baseline = career_t20[career_t20['Matches'] >= 20].copy()

print(f"📊 Baseline: {len(baseline)} bowlers (≥20 matches)")

# ============================================================================
# HELPER FUNCTION: DISPLAY TOP 20
# ============================================================================

def show_top_20(df, approach_name, sort_column='Career_BPI'):
    """Display top 20 bowlers with legend markers"""
    
    # Known legends (adjust names as they appear in data)
    LEGENDS = ['jj bumrah', 'rashid khan', 'sunil narine', 'lasith malinga', 
               'sl malinga', 'r ashwin', 'imran tahir', 'mitchell starc',
               'yuzvendra chahal', 'trent boult']
    
    print(f"\n{'='*80}")
    print(f"📊 {approach_name}")
    print(f"{'='*80}")
    
    top_20 = df.nlargest(20, sort_column).reset_index(drop=True)
    
    print(f"\n{'Rank':<6} {'Bowler':<25} {'BPI':<8} {'Matches':<10} {'Wickets':<10}")
    print(f"{'-'*80}")
    
    legend_count_top20 = 0
    
    for idx, row in top_20.iterrows():
        is_legend = any(legend in row['bowler'].lower() for legend in LEGENDS)
        marker = "⭐" if is_legend else "  "
        
        if is_legend:
            legend_count_top20 += 1
        
        print(f"{idx+1:<6} {row['bowler'][:24]:<25} {row[sort_column]:<8.2f} "
              f"{int(row['Matches']):<10} {int(row['Total_Wickets']):<10} {marker}")
    
    # Legend check
    print(f"\n📊 Legend Analysis:")
    print(f"   Legends in Top 20: {legend_count_top20}/20")
    
    # Find legend rankings
    print(f"\n   Individual Legend Rankings:")
    for legend in LEGENDS:
        legend_data = df[df['bowler'].str.contains(legend, case=False, na=False)]
        if len(legend_data) > 0:
            best = legend_data.nlargest(1, sort_column).iloc[0]
            rank = (df[sort_column] > best[sort_column]).sum() + 1
            status = "✅" if rank <= 20 else "⚠️"
            print(f"      {legend:.<30} Rank: {rank:>4}  BPI: {best[sort_column]:>6.2f}  {status}")

# ============================================================================
# BASELINE APPROACH (ORIGINAL - 20 MATCHES)
# ============================================================================

show_top_20(baseline, "BASELINE: Original BPI (≥20 matches)")

# ============================================================================
# APPROACH 1: HIGHER MATCH THRESHOLD
# ============================================================================

print("\n\n" + "="*80)
print("🔧 APPROACH 1: HIGHER MATCH THRESHOLD")
print("="*80)
print("\nLogic: Require 50 matches minimum instead of 20")
print("Why: Eliminates small-sample flukes")

approach1 = career_t20[career_t20['Matches'] >= 50].copy()
print(f"\n📊 Bowlers remaining: {len(approach1)} (down from {len(baseline)})")

show_top_20(approach1, "APPROACH 1: Higher Threshold (≥50 matches)")

# ============================================================================
# APPROACH 2: OPPOSITION-ADJUSTED BPI
# ============================================================================

print("\n\n" + "="*80)
print("🔧 APPROACH 2: OPPOSITION-ADJUSTED BPI")
print("="*80)
print("\nLogic: Adjust BPI based on opponent strength")
print("Formula: BPI_adj = BPI × (opponent_strength / avg_strength)")

# Calculate opponent strength factor
avg_opponent_strength = baseline['Avg_Opponent_Strength'].mean()

approach2 = baseline.copy()
approach2['Opposition_Factor'] = approach2['Avg_Opponent_Strength'] / avg_opponent_strength
approach2['BPI_OppAdj'] = approach2['Career_BPI'] * approach2['Opposition_Factor']

print(f"\n📊 Opposition adjustment range: {approach2['Opposition_Factor'].min():.3f} - {approach2['Opposition_Factor'].max():.3f}")
print(f"   Factor > 1.0: Bowled against stronger opponents (BPI increases)")
print(f"   Factor < 1.0: Bowled against weaker opponents (BPI decreases)")

show_top_20(approach2, "APPROACH 2: Opposition-Adjusted BPI", sort_column='BPI_OppAdj')

# ============================================================================
# APPROACH 3: CONFIDENCE-WEIGHTED BPI
# ============================================================================

print("\n\n" + "="*80)
print("🔧 APPROACH 3: CONFIDENCE-WEIGHTED BPI")
print("="*80)
print("\nLogic: Weight BPI by statistical confidence")
print("Formula: BPI_conf = BPI × confidence_factor")
print("         confidence_factor = 1 - (1 / sqrt(matches))")

approach3 = baseline.copy()

# Calculate confidence factor
approach3['Confidence_Factor'] = 1 - (1 / np.sqrt(approach3['Matches']))
approach3['BPI_Confidence'] = approach3['Career_BPI'] * approach3['Confidence_Factor']

print(f"\n📊 Confidence factor range: {approach3['Confidence_Factor'].min():.3f} - {approach3['Confidence_Factor'].max():.3f}")
print(f"   20 matches: confidence = {1 - (1/np.sqrt(20)):.3f}")
print(f"   50 matches: confidence = {1 - (1/np.sqrt(50)):.3f}")
print(f"   100 matches: confidence = {1 - (1/np.sqrt(100)):.3f}")

show_top_20(approach3, "APPROACH 3: Confidence-Weighted BPI", sort_column='BPI_Confidence')

# ============================================================================
# APPROACH 4: FULL MEMBER NATIONS ONLY
# ============================================================================

print("\n\n" + "="*80)
print("🔧 APPROACH 4: FULL MEMBER NATIONS FILTER")
print("="*80)
print("\nLogic: Only include bowlers who played ≥50% matches vs Full Members")
print("Why: Focuses on international-level competition")

approach4 = baseline[baseline['Pct_vs_FullMembers'] >= 0.5].copy()

print(f"\n📊 Bowlers remaining: {len(approach4)} (down from {len(baseline)})")
print(f"   Average % vs Full Members: {approach4['Pct_vs_FullMembers'].mean()*100:.1f}%")

show_top_20(approach4, "APPROACH 4: Full Member Filter (≥50% vs Full Members)")

# ============================================================================
# APPROACH 5: RECENCY-WEIGHTED BPI
# ============================================================================

print("\n\n" + "="*80)
print("🔧 APPROACH 5: RECENCY-WEIGHTED BPI")
print("="*80)
print("\nLogic: Weight recent performances higher (exponential decay)")
print("Why: T20 cricket evolved; modern era is harder")

# Need match-level data for this
t20_with_recency = t20_data.copy()

# Calculate years ago
most_recent = t20_with_recency['date'].max()
t20_with_recency['years_ago'] = (most_recent - t20_with_recency['date']).dt.days / 365.25

# Exponential decay weight (decay rate = 0.1)
t20_with_recency['recency_weight'] = np.exp(-0.1 * t20_with_recency['years_ago'])

# Calculate weighted average BPI
approach5 = t20_with_recency.groupby('bowler').apply(
    lambda x: pd.Series({
        'Career_BPI': x['BPI'].mean(),
        'BPI_Recency': np.average(x['BPI'], weights=x['recency_weight']),
        'Matches': len(x),
        'Total_Wickets': x['wickets'].sum(),
        'Avg_Economy': x['economy_rate'].mean(),
        'Avg_Strike_Rate': x['strike_rate'].mean(),
        'Avg_Dot_Pct': x['dot_ball_percentage'].mean(),
        'Career_Start': x['date'].min(),
        'Career_End': x['date'].max(),
        'Pct_vs_FullMembers': x['opponent_is_full_member'].mean(),
        'Avg_Opponent_Strength': x['opponent_batting_strength'].mean()
    })
).reset_index()

approach5 = approach5[approach5['Matches'] >= 20].copy()

print(f"\n📊 Recency weighting:")
print(f"   Match 1 year ago: weight = {np.exp(-0.1 * 1):.3f}")
print(f"   Match 5 years ago: weight = {np.exp(-0.1 * 5):.3f}")
print(f"   Match 10 years ago: weight = {np.exp(-0.1 * 10):.3f}")

show_top_20(approach5, "APPROACH 5: Recency-Weighted BPI", sort_column='BPI_Recency')

# ============================================================================
# APPROACH 6: COMBINED (ALL FIXES)
# ============================================================================

print("\n\n" + "="*80)
print("🔧 APPROACH 6: COMBINED APPROACH")
print("="*80)
print("\nLogic: Apply ALL fixes together")
print("   1. Minimum 40 matches (middle ground)")
print("   2. Opposition adjustment")
print("   3. Confidence weighting")
print("   4. ≥50% vs Full Members")
print("   5. Recency weighting")

# Start with recency-weighted data (already calculated)
approach6 = approach5.copy()

# Fix 1: 40 match minimum
approach6 = approach6[approach6['Matches'] >= 40].copy()

# Fix 2: Opposition adjustment
avg_opp_str = approach6['Avg_Opponent_Strength'].mean()
approach6['Opposition_Factor'] = approach6['Avg_Opponent_Strength'] / avg_opp_str

# Fix 3: Confidence weighting
approach6['Confidence_Factor'] = 1 - (1 / np.sqrt(approach6['Matches']))

# Fix 4: Full Member filter
approach6 = approach6[approach6['Pct_vs_FullMembers'] >= 0.5].copy()

# Fix 5: Already using recency-weighted BPI

# Combine all adjustments
approach6['BPI_Final'] = (
    approach6['BPI_Recency'] * 
    approach6['Opposition_Factor'] * 
    approach6['Confidence_Factor']
)

print(f"\n📊 Filters applied:")
print(f"   Starting bowlers: {len(approach5)}")
print(f"   After ≥40 matches: {(approach5['Matches'] >= 40).sum()}")
print(f"   After Full Member filter: {len(approach6)}")

show_top_20(approach6, "APPROACH 6: COMBINED (All Fixes)", sort_column='BPI_Final')

# ============================================================================
# COMPARISON SUMMARY
# ============================================================================

print("\n\n" + "="*80)
print("📊 APPROACH COMPARISON SUMMARY")
print("="*80)

# Count legends in top 20 for each approach
approaches = {
    'Baseline (≥20)': baseline['Career_BPI'],
    'Approach 1 (≥50)': approach1['Career_BPI'] if len(approach1) > 0 else pd.Series(),
    'Approach 2 (Opp)': approach2['BPI_OppAdj'],
    'Approach 3 (Conf)': approach3['BPI_Confidence'],
    'Approach 4 (Full)': approach4['Career_BPI'] if len(approach4) > 0 else pd.Series(),
    'Approach 5 (Rec)': approach5['BPI_Recency'],
    'Approach 6 (COMB)': approach6['BPI_Final'] if len(approach6) > 0 else pd.Series()
}

LEGENDS = ['jj bumrah', 'rashid khan', 'sunil narine', 'lasith malinga', 
           'sl malinga', 'r ashwin', 'imran tahir', 'mitchell starc']

print(f"\n{'Approach':<20} {'Bowlers':<12} {'Legends in Top 20':<20} {'Best Legend Rank'}")
print(f"{'-'*80}")

for name, (df_name, series) in zip(
    ['Baseline', 'Approach 1', 'Approach 2', 'Approach 3', 'Approach 4', 'Approach 5', 'Approach 6'],
    [
        ('baseline', baseline['Career_BPI']),
        ('approach1', approach1['Career_BPI'] if len(approach1) > 0 else pd.Series()),
        ('approach2', approach2['BPI_OppAdj']),
        ('approach3', approach3['BPI_Confidence']),
        ('approach4', approach4['Career_BPI'] if len(approach4) > 0 else pd.Series()),
        ('approach5', approach5['BPI_Recency']),
        ('approach6', approach6['BPI_Final'] if len(approach6) > 0 else pd.Series())
    ]
):
    if len(series) == 0:
        print(f"{name:<20} {'N/A':<12} {'N/A':<20} {'N/A'}")
        continue
    
    # Get the dataframe
    if df_name == 'baseline':
        df = baseline
    elif df_name == 'approach1':
        df = approach1
    elif df_name == 'approach2':
        df = approach2
        series_col = 'BPI_OppAdj'
    elif df_name == 'approach3':
        df = approach3
        series_col = 'BPI_Confidence'
    elif df_name == 'approach4':
        df = approach4
    elif df_name == 'approach5':
        df = approach5
        series_col = 'BPI_Recency'
    elif df_name == 'approach6':
        df = approach6
        series_col = 'BPI_Final'
    
    # Determine column name
    if df_name in ['baseline', 'approach1', 'approach4']:
        series_col = 'Career_BPI'
    
    # Top 20
    top_20 = df.nlargest(20, series_col)
    
    # Count legends in top 20
    legend_count = sum(
        any(legend in bowler.lower() for legend in LEGENDS)
        for bowler in top_20['bowler']
    )
    
    # Find best legend rank
    best_rank = 999
    for legend in LEGENDS:
        legend_data = df[df['bowler'].str.contains(legend, case=False, na=False)]
        if len(legend_data) > 0:
            best = legend_data.nlargest(1, series_col).iloc[0]
            rank = (df[series_col] > best[series_col]).sum() + 1
            best_rank = min(best_rank, rank)
    
    print(f"{name:<20} {len(df):<12} {legend_count}/20 ({legend_count/20*100:.0f}%){' ':<8} {best_rank}")

print("\n💡 INTERPRETATION:")
print("   Higher 'Legends in Top 20' = Better approach")
print("   Lower 'Best Legend Rank' = Better approach")
print("   Both metrics should improve from Baseline")

print("\n" + "="*80)
print("✅ TESTING COMPLETE!")
print("="*80)

print("\n🎯 NEXT STEP:")
print("   Review results and choose the best approach")
print("   Then apply to ODI and Test formats")
print("   Use refined BPI for modeling")

print("\n" + "="*80)
```

```python
"""
================================================================================
OPPOSITION-ADJUSTED BOWLING PERFORMANCE INDEX (BPI)
================================================================================

Refinement Strategy: Approaches 2 + 4 Combined
- Opposition Adjustment: Weight BPI by opponent quality
- Full Member Filter: Focus on international-level competition

Applied to: All formats (T20, ODI, Test)

================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("\n" + "="*80)
print("🏏 OPPOSITION-ADJUSTED BOWLING PERFORMANCE INDEX")
print("="*80)

# ============================================================================
# STEP 0: LOAD DATA
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 0: LOADING DATA")
print("-"*80)

# 👇 CHANGE THIS PATH
FILE_PATH = 'C:/Users/HPP/abt_with_bpi.parquet'

abt = pd.read_parquet(FILE_PATH)
print(f"✅ Loaded: {len(abt):,} rows with BPI already calculated")

print(f"\n📊 Data by format:")
for fmt in ['T20', 'ODI', 'Test']:
    count = (abt['match_type'] == fmt).sum()
    print(f"   {fmt}: {count:,} performances")

# ============================================================================
# STEP 1: CALCULATE CAREER STATISTICS
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 1: CALCULATING CAREER STATISTICS")
print("-"*80)

career_stats = abt.groupby(['bowler', 'match_type']).agg({
    'BPI': 'mean',
    'wickets': 'sum',
    'economy_rate': 'mean',
    'strike_rate': 'mean',
    'dot_ball_percentage': 'mean',
    'match_id': 'count',
    'opponent_is_full_member': 'mean',
    'opponent_batting_strength': 'mean',
    'bowling_team_won': 'mean',
    'date': ['min', 'max']
}).reset_index()

career_stats.columns = [
    'bowler', 'format', 'Career_BPI_Raw', 'Total_Wickets', 
    'Avg_Economy', 'Avg_Strike_Rate', 'Avg_Dot_Pct', 'Matches',
    'Pct_vs_FullMembers', 'Avg_Opponent_Strength', 'Win_Rate',
    'Career_Start', 'Career_End'
]

print(f"✅ Calculated stats for {len(career_stats)} bowler-format combinations")

# ============================================================================
# STEP 2: APPLY OPPOSITION ADJUSTMENT
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 2: OPPOSITION ADJUSTMENT")
print("-"*80)

print("\n📊 Formula:")
print("   BPI_adjusted = BPI_raw × (opponent_strength / avg_opponent_strength)")
print("\n   Where opponent_strength = average batting strength of teams faced")

# Calculate adjustment factors by format
for fmt in ['T20', 'ODI', 'Test']:
    fmt_data = career_stats[career_stats['format'] == fmt].copy()
    
    if len(fmt_data) == 0:
        continue
    
    # Calculate average opponent strength for this format
    avg_opp_strength = fmt_data['Avg_Opponent_Strength'].mean()
    
    # Calculate opposition factor
    fmt_data['Opposition_Factor'] = fmt_data['Avg_Opponent_Strength'] / avg_opp_strength
    
    # Apply adjustment
    fmt_data['Career_BPI_OppAdj'] = fmt_data['Career_BPI_Raw'] * fmt_data['Opposition_Factor']
    
    # Update back to main dataframe
    career_stats.loc[fmt_data.index, 'Opposition_Factor'] = fmt_data['Opposition_Factor']
    career_stats.loc[fmt_data.index, 'Career_BPI_OppAdj'] = fmt_data['Career_BPI_OppAdj']
    
    print(f"\n   {fmt}:")
    print(f"      Average opponent strength: {avg_opp_strength:.3f}")
    print(f"      Opposition factor range: {fmt_data['Opposition_Factor'].min():.3f} - {fmt_data['Opposition_Factor'].max():.3f}")
    print(f"      Bowlers with factor > 1.0: {(fmt_data['Opposition_Factor'] > 1.0).sum()} "
          f"(faced stronger opponents)")
    print(f"      Bowlers with factor < 1.0: {(fmt_data['Opposition_Factor'] < 1.0).sum()} "
          f"(faced weaker opponents)")

print("\n✅ Opposition adjustment complete!")

# ============================================================================
# STEP 3: APPLY FULL MEMBER FILTER
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 3: FULL MEMBER FILTER")
print("-"*80)

print("\n📊 Filter criteria:")
print("   1. Minimum matches: 20 (baseline)")
print("   2. Played ≥50% matches vs Full Member nations")
print("   3. This focuses on international-level competition")

# Apply filters
career_filtered = career_stats[
    (career_stats['Matches'] >= 20) & 
    (career_stats['Pct_vs_FullMembers'] >= 0.5)
].copy()

print(f"\n📊 Filtering results:")
print(f"   Before filtering: {len(career_stats)} bowler-format combinations")
print(f"   After ≥20 matches: {(career_stats['Matches'] >= 20).sum()}")
print(f"   After Full Member filter: {len(career_filtered)}")
print(f"   Reduction: {len(career_stats) - len(career_filtered)} combinations removed")

print(f"\n📊 By format:")
for fmt in ['T20', 'ODI', 'Test']:
    before = len(career_stats[career_stats['format'] == fmt])
    after = len(career_filtered[career_filtered['format'] == fmt])
    print(f"   {fmt}: {before} → {after} ({after/before*100:.1f}% retained)")

print("\n✅ Filtering complete!")

# ============================================================================
# STEP 4: RANKINGS & VALIDATION
# ============================================================================

print("\n" + "="*80)
print("🏆 STEP 4: LEGEND VALIDATION - OPPOSITION-ADJUSTED BPI")
print("="*80)

# Known legends by format
LEGENDS = {
    'T20': ['jj bumrah', 'rashid khan', 'sunil narine', 'lasith malinga', 
            'sl malinga', 'r ashwin', 'imran tahir', 'mitchell starc',
            'yuzvendra chahal', 'trent boult', 'jasprit bumrah'],
    'ODI': ['m muralitharan', 'wasim akram', 'glenn mcgrath', 'shahid afridi',
            'jj bumrah', 'mitchell starc', 'saqlain mushtaq', 'brett lee',
            'muttiah muralitharan'],
    'Test': ['m muralitharan', 'shane warne', 'anil kumble', 'james anderson',
             'r ashwin', 'dale steyn', 'muttiah muralitharan', 'sk warne',
             'curtly ambrose', 'glenn mcgrath']
}

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt.upper()} FORMAT - TOP 20 BOWLERS (OPPOSITION-ADJUSTED BPI)")
    print(f"{'─'*80}")
    
    fmt_data = career_filtered[career_filtered['format'] == fmt].copy()
    
    if len(fmt_data) == 0:
        print(f"   No data available")
        continue
    
    # Sort by opposition-adjusted BPI
    top_20 = fmt_data.nlargest(20, 'Career_BPI_OppAdj').reset_index(drop=True)
    
    print(f"\n{'Rank':<6} {'Bowler':<30} {'BPI_Adj':<10} {'Raw_BPI':<10} {'Opp_Factor':<12} {'Matches':<10}")
    print(f"{'-'*100}")
    
    legend_count = 0
    
    for idx, row in top_20.iterrows():
        is_legend = any(legend in row['bowler'].lower() for legend in LEGENDS.get(fmt, []))
        marker = "⭐" if is_legend else "  "
        
        if is_legend:
            legend_count += 1
        
        print(f"{idx+1:<6} {row['bowler'][:29]:<30} {row['Career_BPI_OppAdj']:<10.2f} "
              f"{row['Career_BPI_Raw']:<10.2f} {row['Opposition_Factor']:<12.3f} "
              f"{int(row['Matches']):<10} {marker}")
    
    # Legend analysis
    print(f"\n   📊 Legend Validation:")
    print(f"      Legends in Top 20: {legend_count}/20 ({legend_count/20*100:.0f}%)")
    
    # Individual legend rankings
    print(f"\n   📊 Individual Legend Rankings:")
    legend_ranks = []
    
    for legend in LEGENDS.get(fmt, []):
        legend_data = fmt_data[fmt_data['bowler'].str.contains(legend, case=False, na=False)]
        if len(legend_data) > 0:
            best = legend_data.nlargest(1, 'Career_BPI_OppAdj').iloc[0]
            rank = (fmt_data['Career_BPI_OppAdj'] > best['Career_BPI_OppAdj']).sum() + 1
            
            status = "✅ TOP 10" if rank <= 10 else "✅ TOP 20" if rank <= 20 else "⚠️  LOWER"
            
            print(f"      {legend:.<35} Rank: {rank:>4}  BPI: {best['Career_BPI_OppAdj']:>6.2f}  {status}")
            legend_ranks.append(rank)
    
    if legend_ranks:
        avg_rank = np.mean(legend_ranks)
        print(f"\n      Average legend rank: {avg_rank:.1f}")

# ============================================================================
# STEP 5: COMPARISON VISUALIZATION
# ============================================================================

print("\n\n" + "-"*80)
print("📊 STEP 5: CREATING VISUALIZATIONS")
print("-"*80)

fig = plt.figure(figsize=(20, 14))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    fmt_data = career_filtered[career_filtered['format'] == fmt].copy()
    
    if len(fmt_data) == 0:
        continue
    
    # Get top 15 for visualization
    top_15 = fmt_data.nlargest(15, 'Career_BPI_OppAdj')
    
    # Plot 1: Top bowlers horizontal bar
    ax1 = fig.add_subplot(gs[idx, 0])
    
    colors = ['#2ecc71' if any(legend in bowler.lower() 
              for legend in LEGENDS.get(fmt, [])) else '#3498db' 
              for bowler in top_15['bowler']]
    
    y_pos = range(len(top_15))
    ax1.barh(y_pos, top_15['Career_BPI_OppAdj'], color=colors, 
             edgecolor='black', alpha=0.8)
    ax1.set_yticks(y_pos)
    ax1.set_yticklabels([name[:25] for name in top_15['bowler']], fontsize=9)
    ax1.set_xlabel('Opposition-Adjusted BPI', fontsize=11, fontweight='bold')
    ax1.set_title(f'{fmt} - Top 15 Bowlers', fontsize=12, fontweight='bold')
    ax1.invert_yaxis()
    ax1.grid(True, alpha=0.3, axis='x')
    
    # Add value labels
    for i, (y, val) in enumerate(zip(y_pos, top_15['Career_BPI_OppAdj'])):
        ax1.text(val + 0.5, y, f'{val:.1f}', va='center', fontsize=8)
    
    # Plot 2: BPI adjustment scatter
    ax2 = fig.add_subplot(gs[idx, 1])
    
    scatter_data = fmt_data.sample(min(200, len(fmt_data)))
    
    ax2.scatter(scatter_data['Career_BPI_Raw'], 
                scatter_data['Career_BPI_OppAdj'],
                alpha=0.5, s=50, c=scatter_data['Opposition_Factor'],
                cmap='RdYlGn', edgecolors='black', linewidth=0.5)
    
    # Add diagonal reference line
    min_val = min(scatter_data['Career_BPI_Raw'].min(), 
                  scatter_data['Career_BPI_OppAdj'].min())
    max_val = max(scatter_data['Career_BPI_Raw'].max(), 
                  scatter_data['Career_BPI_OppAdj'].max())
    ax2.plot([min_val, max_val], [min_val, max_val], 'r--', 
             linewidth=2, alpha=0.5, label='No adjustment')
    
    ax2.set_xlabel('Raw BPI', fontsize=11, fontweight='bold')
    ax2.set_ylabel('Opposition-Adjusted BPI', fontsize=11, fontweight='bold')
    ax2.set_title(f'{fmt} - BPI Adjustment Impact', fontsize=12, fontweight='bold')
    ax2.legend(fontsize=9)
    ax2.grid(True, alpha=0.3)
    
    # Add colorbar
    sm = plt.cm.ScalarMappable(cmap='RdYlGn', 
                                norm=plt.Normalize(vmin=scatter_data['Opposition_Factor'].min(),
                                                  vmax=scatter_data['Opposition_Factor'].max()))
    sm.set_array([])
    cbar = plt.colorbar(sm, ax=ax2)
    cbar.set_label('Opposition Factor', fontsize=9)
    
    # Plot 3: Opposition factor distribution
    ax3 = fig.add_subplot(gs[idx, 2])
    
    ax3.hist(fmt_data['Opposition_Factor'], bins=30, edgecolor='black', 
             alpha=0.7, color='steelblue')
    ax3.axvline(1.0, color='red', linestyle='--', linewidth=2, 
                label='No adjustment (1.0)')
    ax3.set_xlabel('Opposition Factor', fontsize=11, fontweight='bold')
    ax3.set_ylabel('Number of Bowlers', fontsize=11, fontweight='bold')
    ax3.set_title(f'{fmt} - Opposition Factor Distribution', 
                  fontsize=12, fontweight='bold')
    ax3.legend(fontsize=9)
    ax3.grid(True, alpha=0.3, axis='y')
    
    # Add annotations
    above_1 = (fmt_data['Opposition_Factor'] > 1.0).sum()
    below_1 = (fmt_data['Opposition_Factor'] < 1.0).sum()
    ax3.text(0.05, 0.95, f'Factor > 1.0: {above_1}\n(Tougher opponents)',
             transform=ax3.transAxes, fontsize=9, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))
    ax3.text(0.05, 0.75, f'Factor < 1.0: {below_1}\n(Easier opponents)',
             transform=ax3.transAxes, fontsize=9, verticalalignment='top',
             bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.5))

fig.suptitle('Opposition-Adjusted BPI - Comprehensive Analysis', 
             fontsize=16, fontweight='bold', y=0.995)

plt.savefig('bpi_opposition_adjusted_complete.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: bpi_opposition_adjusted_complete.png")
plt.show()

# ============================================================================
# STEP 6: BEFORE/AFTER COMPARISON
# ============================================================================

print("\n\n" + "-"*80)
print("📊 STEP 6: BEFORE/AFTER COMPARISON")
print("-"*80)

print("\n📊 Comparing Raw BPI vs Opposition-Adjusted BPI:")

for fmt in ['T20', 'ODI', 'Test']:
    fmt_data = career_filtered[career_filtered['format'] == fmt].copy()
    
    if len(fmt_data) == 0:
        continue
    
    print(f"\n{'─'*80}")
    print(f"   {fmt.upper()} FORMAT")
    print(f"{'─'*80}")
    
    # Count legends in top 20
    top_20_raw = fmt_data.nlargest(20, 'Career_BPI_Raw')
    top_20_adj = fmt_data.nlargest(20, 'Career_BPI_OppAdj')
    
    legends_raw = sum(any(legend in bowler.lower() 
                         for legend in LEGENDS.get(fmt, []))
                     for bowler in top_20_raw['bowler'])
    
    legends_adj = sum(any(legend in bowler.lower() 
                         for legend in LEGENDS.get(fmt, []))
                     for bowler in top_20_adj['bowler'])
    
    print(f"   Legends in Top 20:")
    print(f"      Raw BPI: {legends_raw}/20 ({legends_raw/20*100:.0f}%)")
    print(f"      Adjusted BPI: {legends_adj}/20 ({legends_adj/20*100:.0f}%)")
    print(f"      Improvement: {'+' if legends_adj > legends_raw else ''}{legends_adj - legends_raw}")
    
    # Show biggest gainers and losers
    fmt_data['BPI_Change'] = fmt_data['Career_BPI_OppAdj'] - fmt_data['Career_BPI_Raw']
    fmt_data['Rank_Change'] = (
        fmt_data['Career_BPI_Raw'].rank(ascending=False) - 
        fmt_data['Career_BPI_OppAdj'].rank(ascending=False)
    )
    
    print(f"\n   📈 Biggest Gainers (improved ranking most):")
    gainers = fmt_data.nlargest(5, 'Rank_Change')
    for _, row in gainers.iterrows():
        print(f"      {row['bowler'][:30]:.<35} Rank improved by {int(row['Rank_Change']):>3} "
              f"(Opp Factor: {row['Opposition_Factor']:.3f})")
    
    print(f"\n   📉 Biggest Losers (dropped ranking most):")
    losers = fmt_data.nsmallest(5, 'Rank_Change')
    for _, row in losers.iterrows():
        print(f"      {row['bowler'][:30]:.<35} Rank dropped by {int(abs(row['Rank_Change'])):>3} "
              f"(Opp Factor: {row['Opposition_Factor']:.3f})")

# ============================================================================
# STEP 7: SAVE RESULTS
# ============================================================================

print("\n\n" + "-"*80)
print("💾 STEP 7: SAVING RESULTS")
print("-"*80)

# Save career rankings
career_filtered.to_csv('career_bpi_opposition_adjusted.csv', index=False)
print(f"✅ Saved: career_bpi_opposition_adjusted.csv ({len(career_filtered)} combinations)")

# Update match-level data with opposition-adjusted BPI
# Create a mapping of bowler+format to opposition factor
opp_factor_map = career_filtered.set_index(['bowler', 'format'])['Opposition_Factor'].to_dict()

# Apply to match-level data
abt['Opposition_Factor'] = abt.apply(
    lambda row: opp_factor_map.get((row['bowler'], row['match_type']), 1.0),
    axis=1
)
abt['BPI_OppAdj'] = abt['BPI'] * abt['Opposition_Factor']

abt.to_parquet('abt_with_bpi_opposition_adjusted.parquet')
print(f"✅ Saved: abt_with_bpi_opposition_adjusted.parquet ({len(abt):,} rows)")

# ============================================================================
# FINAL SUMMARY
# ============================================================================

print("\n\n" + "="*80)
print("✅ OPPOSITION-ADJUSTED BPI COMPLETE!")
print("="*80)

print("\n🎯 WHAT WAS DONE:")
print("   1. Calculated opposition strength for each bowler")
print("   2. Adjusted BPI based on opponent quality")
print("   3. Filtered to bowlers with ≥50% matches vs Full Members")
print("   4. Validated against cricket legends")
print("   5. Created comprehensive visualizations")

print("\n📊 KEY IMPROVEMENTS:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_data = career_filtered[career_filtered['format'] == fmt]
    if len(fmt_data) == 0:
        continue
    
    top_20 = fmt_data.nlargest(20, 'Career_BPI_OppAdj')
    legend_count = sum(any(legend in bowler.lower() 
                          for legend in LEGENDS.get(fmt, []))
                      for bowler in top_20['bowler'])
    
    print(f"   {fmt}: {legend_count}/20 legends in Top 20 ({legend_count/20*100:.0f}%)")

print("\n📁 FILES CREATED:")
print("   1. career_bpi_opposition_adjusted.csv - Career rankings")
print("   2. abt_with_bpi_opposition_adjusted.parquet - Match-level data")
print("   3. bpi_opposition_adjusted_complete.png - Visualizations")

print("\n🚀 NEXT STEPS:")
print("   1. Use BPI_OppAdj as target for modeling")
print("   2. Predict opposition-adjusted performance")
print("   3. Feature importance analysis")
print("   4. Build format-specific predictive models")
print("   5. SHAP analysis for interpretability")

print("\n" + "="*80)
```

```python
"""
================================================================================
🏏 BOWLING PERFORMANCE INDEX (BPI) - ROW-WISE CALCULATION
================================================================================

Implements:
- Format-specific normalization (0-100)
- Raw BPI calculation
- Opposition adjustment (bowler-format level)
- Hybrid final BPI

Output:
- abt_with_bpi_final_rowwise.parquet
================================================================================
"""

import pandas as pd
import numpy as np

# ============================================================================
# STEP 0: LOAD DATA
# ============================================================================

print("\n📊 Loading ABT...")
FILE_PATH = r"C:\Users\HPP\abt_final_with_all_features.parquet"
abt = pd.read_parquet(FILE_PATH)

print(f"✅ Loaded: {len(abt):,} rows")

# ============================================================================
# STEP 1: FILTER VALID ROWS (wickets ≥ 1)
# ============================================================================

abt = abt[abt['wickets'] >= 1].copy()
print(f"✅ After filtering wickets ≥ 1: {len(abt):,} rows")

# ============================================================================
# STEP 2: NORMALIZATION FUNCTION
# ============================================================================

def normalize_metric(series, lower_is_better=False):
    """Normalize to 0-100 scale with clipping"""
    
    lower = series.quantile(0.01)
    upper = series.quantile(0.99)
    
    clipped = series.clip(lower, upper)
    
    normalized = (clipped - clipped.min()) / (clipped.max() - clipped.min())
    
    if lower_is_better:
        normalized = 1 - normalized
        
    return normalized * 100

# ============================================================================
# STEP 3: FORMAT-SPECIFIC NORMALIZATION
# ============================================================================

print("\n📊 Performing format-wise normalization...")

for fmt in ['T20', 'ODI', 'Test']:
    
    mask = abt['match_type'] == fmt
    df = abt.loc[mask]
    
    print(f"   Processing {fmt}: {len(df)} rows")
    
    # Economy (lower better)
    abt.loc[mask, f'economy_score_{fmt}'] = normalize_metric(
        df['economy_rate'], lower_is_better=True
    )
    
    # Wickets (higher better)
    abt.loc[mask, f'wicket_score_{fmt}'] = normalize_metric(
        df['wickets'], lower_is_better=False
    )
    
    # Strike rate (lower better)
    abt.loc[mask, f'strike_score_{fmt}'] = normalize_metric(
        df['strike_rate'], lower_is_better=True
    )
    
    # Dot ball % (higher better)
    abt.loc[mask, f'dot_score_{fmt}'] = normalize_metric(
        df['dot_ball_percentage'], lower_is_better=False
    )

# ============================================================================
# STEP 4: CALCULATE BPI_RAW
# ============================================================================

print("\n📊 Calculating BPI_Raw...")

abt['BPI_Raw'] = np.nan

# T20
mask = abt['match_type'] == 'T20'
abt.loc[mask, 'BPI_Raw'] = (
    0.287 * abt.loc[mask, 'economy_score_T20'] +
    0.248 * abt.loc[mask, 'wicket_score_T20'] +
    0.245 * abt.loc[mask, 'dot_score_T20'] +
    0.220 * abt.loc[mask, 'strike_score_T20']
)

# ODI
mask = abt['match_type'] == 'ODI'
abt.loc[mask, 'BPI_Raw'] = (
    0.290 * abt.loc[mask, 'strike_score_ODI'] +
    0.270 * abt.loc[mask, 'economy_score_ODI'] +
    0.256 * abt.loc[mask, 'wicket_score_ODI'] +
    0.184 * abt.loc[mask, 'dot_score_ODI']
)

# Test
mask = abt['match_type'] == 'Test'
abt.loc[mask, 'BPI_Raw'] = (
    0.336 * abt.loc[mask, 'wicket_score_Test'] +
    0.279 * abt.loc[mask, 'strike_score_Test'] +
    0.220 * abt.loc[mask, 'economy_score_Test'] +
    0.165 * abt.loc[mask, 'dot_score_Test']
)

# ============================================================================
# STEP 5: OPPOSITION FACTOR (BOWLER-FORMAT LEVEL)
# ============================================================================

print("\n📊 Calculating Opposition Factors...")

# 5a: Avg opponent strength per bowler-format
career_opp = (
    abt.groupby(['bowler', 'match_type'])
    .agg(
        Avg_Opponent_Strength=('opponent_batting_strength', 'mean')
    )
    .reset_index()
)

# 5b: Format average
career_opp['Format_Avg'] = career_opp.groupby('match_type')['Avg_Opponent_Strength'].transform('mean')

# 5c: Opposition factor
career_opp['Opposition_Factor'] = (
    career_opp['Avg_Opponent_Strength'] / career_opp['Format_Avg']
)

# 5d: Cap factor
career_opp['Opposition_Factor_Capped'] = np.clip(
    career_opp['Opposition_Factor'], 0.85, 1.15
)

# Merge back to ABT
abt = abt.merge(
    career_opp[['bowler', 'match_type', 'Avg_Opponent_Strength',
                'Opposition_Factor', 'Opposition_Factor_Capped']],
    on=['bowler', 'match_type'],
    how='left'
)

# ============================================================================
# STEP 6: APPLY OPPOSITION ADJUSTMENT
# ============================================================================

abt['BPI_OppAdj_Capped'] = abt['BPI_Raw'] * abt['Opposition_Factor_Capped']

# ============================================================================
# STEP 7: FINAL BPI (HYBRID)
# ============================================================================

abt['BPI_Final'] = (
    0.65 * abt['BPI_Raw'] +
    0.35 * abt['BPI_OppAdj_Capped']
)

# ============================================================================
# STEP 8: VALIDATION
# ============================================================================

print("\n📊 VALIDATION CHECKS")

print("\nBPI Range:")
print(f"Min: {abt['BPI_Final'].min():.2f}")
print(f"Max: {abt['BPI_Final'].max():.2f}")

print("\nMean BPI by format:")
print(abt.groupby('match_type')['BPI_Final'].mean())

print("\nOpposition Factor Range:")
print(f"Min: {abt['Opposition_Factor_Capped'].min():.3f}")
print(f"Max: {abt['Opposition_Factor_Capped'].max():.3f}")

# ============================================================================
# STEP 9: TOP PERFORMANCES
# ============================================================================

print("\n🏆 TOP 10 PERFORMANCES BY FORMAT")

for fmt in ['T20', 'ODI', 'Test']:
    
    print(f"\n{'='*60}")
    print(f"{fmt} TOP 10")
    print(f"{'='*60}")
    
    top = abt[abt['match_type'] == fmt].nlargest(10, 'BPI_Final')
    
    print(top[['bowler', 'date', 'BPI_Final', 'BPI_Raw', 'wickets',
               'economy_rate', 'strike_rate', 'dot_ball_percentage']])

# ============================================================================
# STEP 10: SAVE OUTPUT
# ============================================================================

OUTPUT_PATH = 'abt_with_bpi_final_rowwise.parquet'
abt.to_parquet(OUTPUT_PATH)

OUTPUT_PATH = 'abt_with_bpi_final_rowwise.csv'
abt.to_csv(OUTPUT_PATH)

print(f"\n💾 Saved: {OUTPUT_PATH}")
print("\n✅ BPI CALCULATION COMPLETE!")
```

```python
"""
================================================================================
ROW-WISE BPI CALCULATION FOR MODELING
================================================================================

Purpose: Calculate Bowling Performance Index (BPI) for each match performance
Target: BPI_Final will be used as the target variable for predictive modeling

Input: abt_final_with_all_features.parquet (77,031 rows)
Output: abt_with_bpi_rowwise.parquet (with BPI columns added)

Methodology:
1. Format-specific normalization (0-100)
2. Format-specific weights (empirically derived)
3. Opposition adjustment (per bowler-format)
4. Hybrid formula: 65% raw + 35% adjusted

================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

print("\n" + "="*80)
print("🎯 ROW-WISE BPI CALCULATION")
print("="*80)

# ============================================================================
# STEP 0: LOAD DATA
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 0: LOADING DATA")
print("-"*80)

# 👇 CHANGE PATH
FILE_PATH = r"C:\Users\HPP\abt_final_with_all_features.parquet"

abt = pd.read_parquet(FILE_PATH)
print(f"✅ Loaded: {len(abt):,} rows × {len(abt.columns)} columns")

# Check required columns exist
required_cols = ['match_type', 'economy_rate', 'wickets', 'strike_rate', 
                 'dot_ball_percentage', 'bowler', 'opponent_batting_strength']

missing = [col for col in required_cols if col not in abt.columns]
if missing:
    print(f"❌ ERROR: Missing columns: {missing}")
    exit()

print(f"\n📊 Data by format:")
for fmt in ['T20', 'ODI', 'Test']:
    count = (abt['match_type'] == fmt).sum()
    print(f"   {fmt}: {count:,} rows")

# Filter to rows with wickets >= 1 (for valid strike_rate)
print(f"\n📊 Filtering to wickets ≥ 1 (for strike_rate validity):")
print(f"   Before: {len(abt):,} rows")
abt_filtered = abt[abt['wickets'] >= 1].copy()
print(f"   After: {len(abt_filtered):,} rows")
print(f"   Removed: {len(abt) - len(abt_filtered):,} rows with 0 wickets")

# ============================================================================
# STEP 1: FORMAT-SPECIFIC WEIGHTS (Empirically Derived)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 1: FORMAT-SPECIFIC WEIGHTS")
print("-"*80)

WEIGHTS = {
    'T20': {
        'economy_rate': 0.287,
        'wickets': 0.248,
        'dot_ball_percentage': 0.245,
        'strike_rate': 0.220
    },
    'ODI': {
        'strike_rate': 0.290,
        'economy_rate': 0.270,
        'wickets': 0.256,
        'dot_ball_percentage': 0.184
    },
    'Test': {
        'wickets': 0.336,
        'strike_rate': 0.279,
        'economy_rate': 0.220,
        'dot_ball_percentage': 0.165
    }
}

print("\n📊 Weights by format:")
for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n   {fmt}:")
    for metric, weight in sorted(WEIGHTS[fmt].items(), key=lambda x: x[1], reverse=True):
        print(f"      {metric:.<30} {weight:.1%}")

# ============================================================================
# STEP 2: NORMALIZE METRICS (0-100 Scale) - FORMAT-SPECIFIC
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 2: NORMALIZING METRICS (0-100 SCALE)")
print("-"*80)

def normalize_to_100(series, lower_is_better=False, percentile_clip=True):
    """
    Normalize a series to 0-100 scale
    
    Parameters:
    - series: pandas Series
    - lower_is_better: True for economy/strike_rate, False for wickets/dots
    - percentile_clip: If True, clip to 1st-99th percentile
    
    Returns: Normalized series (0-100)
    """
    if percentile_clip:
        lower_bound = series.quantile(0.01)
        upper_bound = series.quantile(0.99)
        series_clipped = series.clip(lower_bound, upper_bound)
    else:
        series_clipped = series.copy()
    
    # Normalize to 0-1
    min_val = series_clipped.min()
    max_val = series_clipped.max()
    
    if max_val == min_val:
        return pd.Series(50, index=series.index)  # All same = neutral 50
    
    normalized = (series_clipped - min_val) / (max_val - min_val)
    
    # Invert if lower is better
    if lower_is_better:
        normalized = 1 - normalized
    
    # Scale to 0-100
    return normalized * 100

# Normalize for each format separately
print("\n📊 Normalizing metrics per format...")

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n   Processing {fmt}...")
    
    fmt_mask = abt_filtered['match_type'] == fmt
    fmt_data = abt_filtered[fmt_mask]
    
    if len(fmt_data) == 0:
        print(f"      ⚠️  No data for {fmt}")
        continue
    
    # Normalize each metric
    # Economy & Strike Rate: Lower is better
    abt_filtered.loc[fmt_mask, 'economy_score'] = normalize_to_100(
        fmt_data['economy_rate'], 
        lower_is_better=True
    )
    
    abt_filtered.loc[fmt_mask, 'strike_score'] = normalize_to_100(
        fmt_data['strike_rate'], 
        lower_is_better=True
    )
    
    # Wickets & Dot%: Higher is better
    abt_filtered.loc[fmt_mask, 'wicket_score'] = normalize_to_100(
        fmt_data['wickets'], 
        lower_is_better=False
    )
    
    abt_filtered.loc[fmt_mask, 'dot_score'] = normalize_to_100(
        fmt_data['dot_ball_percentage'], 
        lower_is_better=False
    )
    
    # Show statistics
    print(f"      Economy score: {abt_filtered.loc[fmt_mask, 'economy_score'].mean():.1f} ± {abt_filtered.loc[fmt_mask, 'economy_score'].std():.1f}")
    print(f"      Wicket score: {abt_filtered.loc[fmt_mask, 'wicket_score'].mean():.1f} ± {abt_filtered.loc[fmt_mask, 'wicket_score'].std():.1f}")
    print(f"      Strike score: {abt_filtered.loc[fmt_mask, 'strike_score'].mean():.1f} ± {abt_filtered.loc[fmt_mask, 'strike_score'].std():.1f}")
    print(f"      Dot score: {abt_filtered.loc[fmt_mask, 'dot_score'].mean():.1f} ± {abt_filtered.loc[fmt_mask, 'dot_score'].std():.1f}")

print("\n✅ Normalization complete!")

# ============================================================================
# STEP 3: CALCULATE RAW BPI (Per Row)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 3: CALCULATING RAW BPI")
print("-"*80)

def calculate_raw_bpi(row):
    """Calculate BPI_Raw for a single row"""
    fmt = row['match_type']
    
    if fmt not in WEIGHTS:
        return np.nan
    
    weights = WEIGHTS[fmt]
    
    bpi_raw = (
        weights['economy_rate'] * row['economy_score'] +
        weights['wickets'] * row['wicket_score'] +
        weights['dot_ball_percentage'] * row['dot_score'] +
        weights['strike_rate'] * row['strike_score']
    )
    
    return bpi_raw

print("\n📊 Calculating BPI_Raw for each row...")
abt_filtered['BPI_Raw'] = abt_filtered.apply(calculate_raw_bpi, axis=1)

print(f"✅ BPI_Raw calculated for {len(abt_filtered):,} rows")

print(f"\n📊 BPI_Raw statistics by format:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_filtered[abt_filtered['match_type'] == fmt]['BPI_Raw']
    if len(fmt_bpi) > 0:
        print(f"   {fmt}: Mean={fmt_bpi.mean():.2f}, Median={fmt_bpi.median():.2f}, "
              f"Std={fmt_bpi.std():.2f}, Range=[{fmt_bpi.min():.2f}, {fmt_bpi.max():.2f}]")

# ============================================================================
# STEP 4: OPPOSITION ADJUSTMENT (Per Bowler-Format)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 4: OPPOSITION ADJUSTMENT")
print("-"*80)

print("\n📊 Calculating average opponent strength per bowler-format...")

# Calculate average opponent strength for each (bowler, match_type) combination
bowler_opp_strength = abt_filtered.groupby(['bowler', 'match_type']).agg({
    'opponent_batting_strength': 'mean'
}).reset_index()

bowler_opp_strength.columns = ['bowler', 'match_type', 'Avg_Opponent_Strength']

print(f"✅ Calculated for {len(bowler_opp_strength)} bowler-format combinations")

# Calculate format-average opponent strength
print("\n📊 Calculating opposition factors by format...")

for fmt in ['T20', 'ODI', 'Test']:
    fmt_data = bowler_opp_strength[bowler_opp_strength['match_type'] == fmt]
    
    if len(fmt_data) == 0:
        continue
    
    # Format average
    format_avg = fmt_data['Avg_Opponent_Strength'].mean()
    
    # Calculate opposition factor
    fmt_mask = bowler_opp_strength['match_type'] == fmt
    bowler_opp_strength.loc[fmt_mask, 'Opposition_Factor'] = (
        bowler_opp_strength.loc[fmt_mask, 'Avg_Opponent_Strength'] / format_avg
    )
    
    # Cap the factor
    bowler_opp_strength.loc[fmt_mask, 'Opposition_Factor_Capped'] = np.clip(
        bowler_opp_strength.loc[fmt_mask, 'Opposition_Factor'],
        0.85,
        1.15
    )
    
    # Statistics
    factors = bowler_opp_strength.loc[fmt_mask, 'Opposition_Factor']
    capped = bowler_opp_strength.loc[fmt_mask, 'Opposition_Factor_Capped']
    
    print(f"\n   {fmt}:")
    print(f"      Format avg opponent strength: {format_avg:.1f}")
    print(f"      Opposition factor range: {factors.min():.3f} - {factors.max():.3f}")
    print(f"      Capped factor range: {capped.min():.3f} - {capped.max():.3f}")
    print(f"      Bowlers capped at 1.15: {(factors > 1.15).sum()}")
    print(f"      Bowlers capped at 0.85: {(factors < 0.85).sum()}")

print("\n✅ Opposition factors calculated!")

# Map opposition factors back to each row
print("\n📊 Mapping opposition factors to each row...")

abt_filtered = abt_filtered.merge(
    bowler_opp_strength[['bowler', 'match_type', 'Opposition_Factor', 'Opposition_Factor_Capped']],
    on=['bowler', 'match_type'],
    how='left'
)

# Fill any missing values with 1.0 (neutral)
abt_filtered['Opposition_Factor'].fillna(1.0, inplace=True)
abt_filtered['Opposition_Factor_Capped'].fillna(1.0, inplace=True)

print(f"✅ Opposition factors mapped to all rows")

# ============================================================================
# STEP 5: CALCULATE OPPOSITION-ADJUSTED BPI (Per Row)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 5: OPPOSITION-ADJUSTED BPI")
print("-"*80)

print("\n📊 Calculating BPI_OppAdj_Capped...")
abt_filtered['BPI_OppAdj_Capped'] = (
    abt_filtered['BPI_Raw'] * abt_filtered['Opposition_Factor_Capped']
)

print(f"\n📊 BPI_OppAdj_Capped statistics by format:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_filtered[abt_filtered['match_type'] == fmt]['BPI_OppAdj_Capped']
    if len(fmt_bpi) > 0:
        print(f"   {fmt}: Mean={fmt_bpi.mean():.2f}, Range=[{fmt_bpi.min():.2f}, {fmt_bpi.max():.2f}]")

print("\n✅ Opposition-adjusted BPI calculated!")

# ============================================================================
# STEP 6: HYBRID FORMULA - FINAL BPI (Per Row)
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 6: FINAL BPI (HYBRID FORMULA)")
print("-"*80)

print("\n📊 Formula: BPI_Final = 0.65 × BPI_Raw + 0.35 × BPI_OppAdj_Capped")

abt_filtered['BPI_Final'] = (
    0.65 * abt_filtered['BPI_Raw'] + 
    0.35 * abt_filtered['BPI_OppAdj_Capped']
)

print(f"\n✅ BPI_Final calculated for {len(abt_filtered):,} rows")

print(f"\n📊 BPI_Final statistics by format:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_filtered[abt_filtered['match_type'] == fmt]['BPI_Final']
    if len(fmt_bpi) > 0:
        print(f"   {fmt}: Mean={fmt_bpi.mean():.2f}, Median={fmt_bpi.median():.2f}, "
              f"Std={fmt_bpi.std():.2f}, Range=[{fmt_bpi.min():.2f}, {fmt_bpi.max():.2f}]")

# ============================================================================
# STEP 7: VALIDATION & QUALITY CHECKS
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 7: VALIDATION CHECKS")
print("-"*80)

# Check for NaN values
print(f"\n📊 Missing value check:")
print(f"   BPI_Raw: {abt_filtered['BPI_Raw'].isna().sum()} missing")
print(f"   BPI_OppAdj_Capped: {abt_filtered['BPI_OppAdj_Capped'].isna().sum()} missing")
print(f"   BPI_Final: {abt_filtered['BPI_Final'].isna().sum()} missing")

# Check ranges
print(f"\n📊 Range validation:")
print(f"   BPI_Final min: {abt_filtered['BPI_Final'].min():.2f} (should be ≥ 0)")
print(f"   BPI_Final max: {abt_filtered['BPI_Final'].max():.2f} (should be ≤ 115)")
print(f"   Opposition_Factor_Capped range: [{abt_filtered['Opposition_Factor_Capped'].min():.3f}, "
      f"{abt_filtered['Opposition_Factor_Capped'].max():.3f}] (should be [0.85, 1.15])")

# Overall statistics
print(f"\n📊 Overall BPI_Final:")
print(f"   Mean: {abt_filtered['BPI_Final'].mean():.2f}")
print(f"   Median: {abt_filtered['BPI_Final'].median():.2f}")
print(f"   Std Dev: {abt_filtered['BPI_Final'].std():.2f}")
print(f"   25th percentile: {abt_filtered['BPI_Final'].quantile(0.25):.2f}")
print(f"   75th percentile: {abt_filtered['BPI_Final'].quantile(0.75):.2f}")

# ============================================================================
# STEP 8: TOP PERFORMANCES BY FORMAT
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 8: TOP 10 PERFORMANCES BY FORMAT")
print("-"*80)

for fmt in ['T20', 'ODI', 'Test']:
    print(f"\n{'─'*80}")
    print(f"🏏 {fmt} - TOP 10 PERFORMANCES")
    print(f"{'─'*80}")
    
    fmt_data = abt_filtered[abt_filtered['match_type'] == fmt].copy()
    top_10 = fmt_data.nlargest(10, 'BPI_Final')
    
    print(f"\n{'Rank':<6} {'Bowler':<25} {'BPI_Final':<12} {'Wickets':<10} {'Economy':<10} {'Date'}")
    print("-"*80)
    
    for idx, (i, row) in enumerate(top_10.iterrows(), 1):
        date_str = str(row['date'])[:10] if 'date' in row else 'N/A'
        print(f"{idx:<6} {row['bowler'][:24]:<25} {row['BPI_Final']:<12.2f} "
              f"{int(row['wickets']):<10} {row['economy_rate']:<10.2f} {date_str}")

# ============================================================================
# STEP 9: VISUALIZATION
# ============================================================================

print("\n" + "-"*80)
print("📊 STEP 9: CREATING VISUALIZATIONS")
print("-"*80)

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('Row-wise BPI - Distribution Analysis', fontsize=16, fontweight='bold')

for idx, fmt in enumerate(['T20', 'ODI', 'Test']):
    fmt_data = abt_filtered[abt_filtered['match_type'] == fmt]
    
    # BPI_Final distribution
    ax = axes[0, idx]
    ax.hist(fmt_data['BPI_Final'], bins=50, edgecolor='black', alpha=0.7, color='steelblue')
    ax.axvline(fmt_data['BPI_Final'].mean(), color='red', linestyle='--', 
               linewidth=2, label=f'Mean: {fmt_data["BPI_Final"].mean():.1f}')
    ax.axvline(fmt_data['BPI_Final'].median(), color='green', linestyle='--',
               linewidth=2, label=f'Median: {fmt_data["BPI_Final"].median():.1f}')
    ax.set_xlabel('BPI_Final', fontsize=11)
    ax.set_ylabel('Frequency', fontsize=11)
    ax.set_title(f'{fmt} - BPI_Final Distribution', fontsize=12, fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    # Raw vs Final BPI
    ax = axes[1, idx]
    sample = fmt_data.sample(min(1000, len(fmt_data)))
    ax.scatter(sample['BPI_Raw'], sample['BPI_Final'], alpha=0.3, s=20, color='coral')
    
    # Add reference line
    min_val = min(sample['BPI_Raw'].min(), sample['BPI_Final'].min())
    max_val = max(sample['BPI_Raw'].max(), sample['BPI_Final'].max())
    ax.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, 
            alpha=0.5, label='No adjustment')
    
    corr = fmt_data['BPI_Raw'].corr(fmt_data['BPI_Final'])
    ax.set_xlabel('BPI_Raw', fontsize=11)
    ax.set_ylabel('BPI_Final', fontsize=11)
    ax.set_title(f'{fmt} - Raw vs Final (r={corr:.3f})', fontsize=12, fontweight='bold')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('bpi_rowwise_analysis.png', dpi=300, bbox_inches='tight')
print(f"\n💾 Saved: bpi_rowwise_analysis.png")
plt.show()

# ============================================================================
# STEP 10: SAVE ENHANCED ABT
# ============================================================================

print("\n" + "-"*80)
print("💾 STEP 10: SAVING ENHANCED ABT")
print("-"*80)

# Save with BPI columns
output_file = 'abt_with_bpi_rowwise.parquet'
abt_filtered.to_parquet(output_file)
print(f"✅ Saved: {output_file}")
print(f"   Rows: {len(abt_filtered):,}")
print(f"   Columns: {len(abt_filtered.columns)} (original {len(abt.columns)} + 9 BPI columns)")

# Also save as CSV for inspection
abt_filtered.to_csv('abt_with_bpi_rowwise.csv', index=False)
print(f"✅ Saved: abt_with_bpi_rowwise.csv (for inspection)")

# List new columns added
new_cols = ['economy_score', 'wicket_score', 'strike_score', 'dot_score',
            'BPI_Raw', 'Opposition_Factor', 'Opposition_Factor_Capped',
            'BPI_OppAdj_Capped', 'BPI_Final']

print(f"\n📊 New columns added:")
for col in new_cols:
    print(f"   ✓ {col}")

# ============================================================================
# FINAL SUMMARY
# ============================================================================

print("\n\n" + "="*80)
print("✅ ROW-WISE BPI CALCULATION COMPLETE!")
print("="*80)

print(f"\n🎯 SUMMARY:")
print(f"   Input: {len(abt):,} rows")
print(f"   Filtered (wickets ≥ 1): {len(abt_filtered):,} rows")
print(f"   New columns added: {len(new_cols)}")

print(f"\n📊 BPI_Final by Format:")
for fmt in ['T20', 'ODI', 'Test']:
    fmt_bpi = abt_filtered[abt_filtered['match_type'] == fmt]['BPI_Final']
    print(f"   {fmt}: {len(fmt_bpi):,} rows, Mean={fmt_bpi.mean():.2f}")

print(f"\n📁 Files Created:")
print(f"   1. abt_with_bpi_rowwise.parquet - Enhanced ABT")
print(f"   2. abt_with_bpi_rowwise.csv - CSV version")
print(f"   3. bpi_rowwise_analysis.png - Visualization")

print(f"\n🚀 READY FOR MODELING:")
print(f"   Target variable: BPI_Final")
print(f"   Data: {len(abt_filtered):,} match-level performances")
print(f"   Features: {len(abt_filtered.columns) - len(new_cols)} original columns")

print(f"\n🎯 NEXT STEPS:")
print(f"   1. Feature selection")
print(f"   2. Train/val/test temporal split (≤2023, 2024, 2025)")
print(f"   3. Format-specific models")
print(f"   4. Feature importance & SHAP analysis")

print("\n" + "="*80)
```
